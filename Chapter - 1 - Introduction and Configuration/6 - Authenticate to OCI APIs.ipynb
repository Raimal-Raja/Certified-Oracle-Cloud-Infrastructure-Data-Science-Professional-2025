### Importance of authentication in Data Science
Data science resources allow you to execute custom code.

as part of your code, you may want to interact with other OCI service  via the OCI Rest APIs.

to  interact with OCI APIs, you'll need to operate as an authenticated user.


### What are resource principals?
resource principals are feature of IAM that enable resources to be authorized principal actors that perform actions on service resources. Each resource:
has its own identity, authenticates by using certficates added



### Resource  principals and data science service authentication
the data science service enables you to authenticate by using your notebook session's or job run's resource principal to access other OCI resources.

### Resource principals
#### source
Provide a more secure way to authenticate resources compared to the oci configuration and API key approaches, 
#### Practical
Make authentication practical and easier for job runs, which don't have an interactive interface for creating and locating, config, flies.

#### Explicit
Require explicit use when invoking, and SDK or CLI (otherwise, the configuration files and API key approach is used)


#### Cached
Tokens are cached for 15 mins, (if you change the policy or dynamic group, you must wait 15 mins to see the changes)
