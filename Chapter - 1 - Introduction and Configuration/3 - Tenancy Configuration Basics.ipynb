### Compartments
Enable you to organize and control access to your cloud resources.

a logical grouping of resources that can be accessed only by certain groups that have recieved administrator permission.

tipl when configuring tenancies, decide how you will organize your data science resources, then create compartments for those resources through the identity console.

#### Creating compartment -cloud console
go to identify and select compartments, click the create compartment button, enter name, description, and click create compartment

### User groups
individual users are grouped in oci and granted access to data science resources within compartments

admins perform three steps to create user groups.
1 - create users, 2 - create groups, 3 - add users to groups,

### Dynamic groups
contain resources such as data science notebook sessions, job runs, and model deplyments. Resources match rules that you define.


Matching rules allow group memberships to change dynamically ad resources that match those rules are created or deleted.

Group resources serve as principal actors and can make api calls to services according to policies.
