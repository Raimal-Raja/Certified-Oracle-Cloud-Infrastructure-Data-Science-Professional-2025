## 2. Understanding Data Visualization (DV)

**Data Visualization (DV)** refers to presenting data graphically to:
- Illustrate findings.
- Explain results.
- Facilitate decision-making.

### Key Benefits:
- Quick and easy interpretation of data.
- No special training needed to read visuals.
- Enables discovery of hidden patterns and correlations.

### Data Source Integration:
Most DV tools can connect to data sources like:
- Relational databases (on-premises or cloud-based).
- APIs and flat files.
- Enterprise data warehouses.

Users can choose from a wide range of visual options, and some tools even **auto-recommend** chart types based on the data.

## 3. Features of an Effective Data Visualization Tool

A good DV tool should:
- **Integrate seamlessly** with multiple data sources.

ADS offers multiple visualization methods to explore and profile data:

| Method | Description |

| `plot()` | Automatic plotting tool for univariate or bivariate analysis. |

## 6. Correlation Visualization (`corr()`)

To visualize correlations, use:
```python
dataset.show_corr()
````

If correlation matrices are not cached, this command triggers `corr()` to calculate them.

### Correlation Types:

* **Continuous–Continuous:** Pearson correlation coefficient
  Range: -1 to +1 → measures linear relationships.
* **Continuous–Categorical:** Correlation ratio (η²)
  Range: 0 to 1 → measures curvilinear relationships.
* **Categorical–Categorical:** Cramer’s V
  Range: 0 to 1 → measures association between nominal variables.

*Note:* Each correlation type is displayed independently since scales differ.

---

## 7. Dataset Preview (`show_in_notebook()`)

This function creates a **comprehensive preview** of a dataset, showing:

* Dataset type (regression, binary, or multiclass classification)
* Number of rows and columns
* Data types for each feature
* Visualizations per feature
* Correlation map
* Sample of data header

### Performance Note:

`show_in_notebook()` uses a **smart sampling method** that is:

* Statistically significant at **95% confidence level**
* Within a **1% confidence interval**

---

## 8. Plotting Relationships (`plot()`)

Use `plot()` to visualize relationships between features.

### Example:

```python
dataset.plot(x="Attrition", y="MonthlyIncome").show_in_notebook()
```

### Automatic Plot Type Detection:

* **Categorical (x only):** Bar chart (e.g., Attrition: 0 or 1)
* **Continuous:** Histogram
* **Categorical vs Continuous:** Violin plot
* **Continuous vs Continuous:** Gaussian heatmap or scatterplot

ADS intelligently determines the **best-fit plot type** based on feature types.

---

## 9. Feature Type System in ADS

The **Feature Type System** in ADS:

* Extends pandas DataFrames.
* Stores metadata, properties, and methods for features.
* Uses **warnings and validation systems** for data quality assurance.
* Separates **data representation** (format) from **data meaning** (measurement).

### Key Capabilities:

* Automatically detects feature types.
* Provides visualizations for supported feature types.
* Allows creation of **custom feature plots** for advanced exploration.

---

## 10. Feature Plots (`feature_plot()`)

* `feature_plot()` on a **pandas Series** → produces a *univariate* plot.
* `feature_plot()` on a **DataFrame** → returns a summary plot for each feature.

This makes it easy to review all features’ visual summaries in one view.

---

## 11. Custom Visualizations

ADS provides flexibility to use **any external visualization library**.

### Example: Using Seaborn

```python
import seaborn as sns
sns.pairplot(dataset.to_pandas())
```

* Creates pairwise feature relationships.
* Diagonal axes show univariate distributions.
* Each feature appears across rows and columns.

### Example: Using Matplotlib

```python
import matplotlib.pyplot as plt
plt.scatter(df['Magnitude'], df['Depth'])
plt.title("California Earthquake Data")
plt.show()
```

This example shows a **public California earthquake dataset**, providing a visual overview of earthquake occurrences.
