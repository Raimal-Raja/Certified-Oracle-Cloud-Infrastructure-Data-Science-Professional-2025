## 🔍 Understanding Serialization

**Serialization** is the process of converting a data object (such as Python objects or TensorFlow models) into a format that can be stored or transmitted and then reconstructed later.  
The reverse process is known as **deserialization**.

### Common Serialization Formats
- **JSON** – Human-readable format, great for text-based data exchange.  
- **XML** – Structured, markup-based data format.  
- **HDF5** – Used for large numerical datasets.  
- **Pickle (Python)** – Converts Python objects into byte arrays for persistence.

Serialization is also referred to as **marshaling** in some programming contexts.

---

## 🧠 Serialization in ADS (Oracle Accelerated Data Science)

The **serialization object** in ADS has various model-specific classes:
- For **PyTorch models** → Use the *PyTorch model class*.  
- For **TensorFlow models** → Use the *TensorFlow model class*.  

While ADS supports many frameworks, it’s not feasible to provide serialization classes for every model type.

## 💾 Saving and Uploading Models

The **save() method** is used to store model artifacts in the model catalog.

### Steps in the Save Process
1. **Reloads** `score.py` and `runtime.yaml` from disk to pick up any modifications.  
2. **Performs introspection tests** (if `ignore_introspection=False`) to detect deployment issues and suggest fixes.  
3. **Uploads artifacts** to the Model Catalog and returns the **model OCID** (Oracle Cloud Identifier).

You can also manually call `introspect()` after `prepare()` to conduct the introspection test again.

## 🧩 Generic Model Preparation

The `prepare_generic_model()` function helps convert **any model type** into an ADS-compatible model object.

ADS assists by:
- Generating **standardized code templates**,  
- Allowing users to **customize** for their use case,  
- And then saving the prepared **model object and metadata** to the catalog.

Models can be saved using three interfaces:
1. **ADS SDK**  
2. **OCI Python SDK**  
3. **OCI Console**

These allow for both programmatic and console-based model management.

## 🗂️ Viewing and Editing Models

Within the **Model Catalog**, users can:
- View model details,  
- Edit metadata (provenance, taxonomy),  
- Move models across compartments,  
- Activate or deactivate models,  
- And delete models if necessary.

### Tags and Provenance
- **Tags Tab** – Apply defined or free-form tags.  
- **Training Resource Tab** – Displays the notebook session or job that trained the model.  
- **Source Code Tab** – Shows Git details, training scripts, and other resources.

---

## 🧾 Model Taxonomy and Introspection

### Model Taxonomy View
Displays:
- Model descriptions and taxonomy,  
- Custom attributes,  
- Associated model deployments.

### Introspection Tests
When defined, these show the results of client-side tests (via `score.py` or `runtime.yaml`), with statuses such as:
- ✅ Success  
- ❌ Failed  
- ⚪ Not Tested  

> It’s recommended that all tests pass before saving a model.

---

## 🧮 Input and Output Schemas

If defined, these schemas describe the structure of input and output data for the model.  
They can be reviewed and copied directly within the model catalog interface.  

Users **cannot edit schemas or the model** itself once saved, but they can update:
- The **model name**,  
- The **description**,  
- And **metadata** (provenance and taxonomy).

---

## 🔄 Moving and Activating Models

Models can be:
- **Moved** between compartments (e.g., from development → production).  
- **Activated** to make them deployable.  
- **Deactivated** to prevent deployment without deletion.

> Inactive models remain stored but cannot be deployed until reactivated.

---

## 🗑️ Deleting Models

To delete a model:
1. Click its name or use the **Actions** menu.  
2. Select **Delete**.  
3. Enter the exact (case-insensitive) model name.  

Deleted models:
- Remain in the catalog for **30 days**.  
- Can be **filtered out** using the *state filter*.

---

## ⚙️ Supported Model Operations

Users can perform the following operations inside the Model Catalog:

| Operation | Description |
|------------|-------------|
| **List models** | View all models in the catalog. |
| **Edit metadata** | Update taxonomy and provenance (not the model code). |
| **Move models** | Transfer models between compartments. |
| **Activate/Deactivate** | Control model deployment availability. |
| **Delete models** | Soft-delete (remains for 30 days). |
| **CLI/SDK Operations** | Create, update, list, and delete models programmatically. |
