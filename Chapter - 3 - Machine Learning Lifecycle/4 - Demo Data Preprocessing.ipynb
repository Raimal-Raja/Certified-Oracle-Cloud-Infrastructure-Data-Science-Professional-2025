
# Hands-On Demo: Data Preprocessing with Oracle Cloud Infrastructure Data Science

## 1. Introduction

In this demo, we will go through a **hands-on example of preprocessing** a dataset using **Oracle Cloud Infrastructure (OCI) Data Science** and the **Accelerated Data Science (ADS)** SDK.

**Dataset Used:** *Employee Attrition Dataset*  
- **Rows:** 1,470  
- **Features:** 36  
  - 22 ordinal features  
  - 11 categorical features  
  - 3 constant features  
- **Includes:** demographic info, compensation levels, job characteristics, job satisfaction, and performance metrics.  
- **Data Imbalance:** fewer employees leave than stay.

---

## 2. Importing Required Libraries

In the **first code cell**, we import all the necessary libraries, including:
- **ADS (Accelerated Data Science)**
- **pandas**
- **NumPy**
- **Other standard data science libraries**

### Loading Dataset from Object Storage
We use the **`resource_principal`** method to access data securely from OCI Object Storage.  
This is defined in the **Setup function**, allowing us to authenticate and access resources programmatically.

### Example:
```python
from ads.dataset.factory import DatasetFactory
from oci.object_storage import ObjectStorageClient
````

---

## 3. Accessing Dataset in OCI Notebook Session

We define:

* **Bucket name**
* **Namespace**
* **Dataset path**

Then, using:

```python
dataset = DatasetFactory.open("oci://<bucket>@<namespace>/<dataset_path>")
```

We load the data into an ADS dataset object.
We also define the **target feature** as `Attrition`.

---

## 4. Using `suggest_recommendations()`

### Function Overview:

`dataset.suggest_recommendations()` analyzes the dataset and shows:

* Detected data issues
* Recommended transformations
* Suggested actions with code examples

### Example Output:

* Messages indicating missing values
* Recommendations for encoding categorical variables
* Suggested actions to handle class imbalance

This feature simplifies identifying preprocessing needs.

---

## 5. Using `auto_transform()`

### Function Overview:

`auto_transform()` applies **all suggested transformations automatically** and returns a **transformed dataset**.

It performs:

* Missing value imputation
* Noise reduction
* Feature correlation removal
* Upsampling/downsampling (for class balance)
* Removal of non-predictive columns (e.g., IDs, constraints)

### Example:

```python
transformed_dataset = dataset.auto_transform()
```

The transformations happen behind the scenes, optimizing data for modeling.

---

## 6. Visualizing Transformations

Use `visualize_transforms()` to **display a flow diagram** showing all transformations performed.

### Comparison:

* **Without Auto Transform:** Shows raw dataset.
* **With Auto Transform:** Displays improved, processed dataset.

This helps visualize the preprocessing workflow and ensures transparency.

---

## 7. Encoding Categorical Data

Let’s explore the **Job Function** feature.

### Step 1: Check Categories

```python
dataset['JobFunction'].value_counts()
```

You’ll see three categorical values representing different job types.

### Step 2: Encode with ADS Label Encoder

ADS provides a built-in encoder under:

```python
from ads.dataset.label_encoder import LabelEncoder
```

Encoding converts text-based categories into numerical labels:

* Before: “Sales”, “HR”, “Research”
* After: 0, 1, 2

This is essential for algorithms that require numerical input.

---

## 8. Handling Imbalanced Data with Upsampling

The dataset is imbalanced — fewer employees leave than stay.

To balance it, we use:

```python
from ads.dataset.helper import upsample
upsampled_data = upsample(dataset, target='Attrition')
```

### After Upsampling:

* The minority class (“Yes”, meaning employee leaves) is duplicated until both classes have equal representation.
* The result: **balanced dataset** for fair model training.

You can confirm the change by comparing `value_counts()` before and after upsampling.

---

## 9. Comparing Attrition Counts

After upsampling:

```python
print(upsampled_data['Attrition'].value_counts())
```

Now, both `Yes` and `No` classes have similar counts — reducing bias during training.

---

## 10. Final Steps — Splitting the Dataset

Once all transformations are done, we **split the dataset** into:

* **Training set**
* **Testing set**
* **Validation set** (optional)

### Default Split in ADS:

* Training: **80%**
* Testing: **10%**
* Validation: **10%**

### Example:

```python
train, test, val = dataset.train_test_val_split()
```

This ensures that model performance can be evaluated on unseen data effectively.

---

## 11. Summary

* We demonstrated **data preprocessing** in OCI Data Science using the **ADS SDK**.
* Covered essential tools:

  * `suggest_recommendations()`
  * `auto_transform()`
  * `visualize_transforms()`
  * `label_encoder`
  * `upsample()`
* Learned to handle missing data, encode categorical features, balance datasets, and split data for training.

### References:

* **ADS Documentation**
* **Oracle GitHub Labs**
* **Data Science Notebook Samples**
```
