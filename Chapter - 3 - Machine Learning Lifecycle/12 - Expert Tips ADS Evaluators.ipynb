## 🧭 The Role of Model Evaluation

Model evaluation occurs **after the training phase** in the machine learning lifecycle.  
Models are only as valuable as their **quality of predictions**.

To measure performance, we use **evaluation metrics** that compare predicted and actual target values.

---

## ⚙️ Evaluation Process

1. **Hold out a validation set** labeled with target features.  
2. **Generate predictions** using your trained ML model.  
3. **Compute evaluation metrics**, which quantify how close predictions are to the true values.

These metrics transform raw model outputs into **interpretable and standardized scores** through charts and statistics.

---

## 🌟 Benefits of Model Evaluation

- **Benchmarking**  
  Helps track and compare performance across models (e.g., examining a confusion matrix).

- **Discovering Pitfalls**  
  Understand issues like *high accuracy but low precision*, or detect random guessing behavior.

- **Understanding Trade-offs**  
  For instance, Model A may perform well in clear weather but poorly in bad conditions.

---

## 🧩 ADS Evaluation Tools

To simplify evaluation and model selection, **Oracle’s ADS** offers the **Evaluation Class**,  
which provides a rich suite of:

- **Tools**
- **Metrics**
- **Charts**

### ADS offers three types of evaluators:

1. **Binary Classification Evaluator**  
   Output: 0 or 1 (e.g., spam detection)

2. **Multiclass Classification Evaluator**  
   Output: Discrete categories (e.g., colors, types)

3. **Regression Evaluator**  
   Output: Continuous values (e.g., price, length)

You can also integrate **open-source libraries** like **scikit-learn** with ADS for clustering and other evaluation tasks.

---

## 🧮 Evaluating Binary Classification Models

### Example Metrics in ADS:
- Accuracy  
- Precision  
- Recall  
- F1 Score  
- ROC/AUC  

### Example Workflow:
```python
from ads.evaluations import ADSEvaluator, ADSModel
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# Convert trained models into ADSModel objects
lr_model = ADSModel.from_estimator(LogisticRegression())
rf_model = ADSModel.from_estimator(RandomForestClassifier())

# Generate evaluation metrics
evaluator = ADSEvaluator([lr_model, rf_model])
evaluator.metrics
````

### Example Output:

* Precision (LogisticRegression): **0.9024**
* Precision (RandomForestClassifier): **0.8814**

### Displaying Charts:

```python
evaluator.show_in_notebook(perfect=True)
```

This displays charts like **Lift & Gain**, **Precision-Recall Curves**, and **Confusion Matrices**.

You can also add custom metrics:

```python
evaluator.add_metrics("F2_Score")
```

---

## 📊 Evaluating Multiclass Classification Models

### Common Metrics:

* Hamming Loss
* F1 Score (macro, micro, weighted)
* Precision and Recall (macro, micro, weighted)
* ROC/AUC

### Example Code:

```python
evaluator = ADSEvaluator.from_estimator(
    estimator=RandomForestClassifier(),
    X=test_data,
    y=test_labels,
    class_levels=[0, 1, 2]
)

evaluator.metrics
```

You can visualize:

* **Multiclass ROC Curves**
* **Precision by Label Charts**
* **F1 by Label Charts**
* **Multiclass PR Curves**

---

## 📈 Evaluating Regression Models

### Example Metrics:

* R² (R-squared)
* Explained Variance
* Mean Squared Error (MSE)
* Mean Absolute Error (MAE)
* Root Mean Squared Error (RMSE)
* Mean Residuals

### Example Code:

```python
evaluator = ADSEvaluator.from_estimator(regression_model, X_test, y_test)
evaluator.metrics
evaluator.show_in_notebook()
```

### Common Regression Charts:

1. **Observed vs Predicted** — plots true vs predicted values.
2. **Residuals QQ Plot** — shows how residuals follow a normal distribution.
3. **Residuals vs Predicted** — visualizes prediction consistency.
4. **Residuals vs Observed** — compares prediction errors with actual data.

---

## 🧠 Key Takeaways

* **ADS Evaluators** streamline model evaluation with **metrics and visualization tools**.
* They support **binary**, **multiclass**, and **regression** models.
* You can use **custom metrics**, **charts**, and **comparisons** between multiple models.
* Integrate with **scikit-learn** and other libraries for extended flexibility.

---
