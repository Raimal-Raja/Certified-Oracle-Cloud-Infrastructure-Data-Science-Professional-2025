## 1. Introduction to AutoML

In this lesson, **Jon Stanesby** explains how to build and train models using **AutoML** within **Accelerated Data Science (ADS)**.  

**AutoML (Automated Machine Learning)** automates the process of model selection, hyperparameter tuning, and evaluation—enabling data scientists to produce optimal models efficiently.

In the modeling phase of the machine learning lifecycle:
- Multiple models are built using different algorithms.
- Each model is tuned with different hyperparameter configurations.
- Performance and accuracy must be tracked and compared.

This entire process can be **automated using AutoML**, which saves time, improves efficiency, and enhances accuracy.

---

## 2. Why AutoML?

Machine learning typically requires **multiple iterations and experiments** to reach the best-performing model. It’s rare to achieve optimal performance on the first attempt.  

AutoML automates this iterative process by:
- Selecting and refining models.  
- Tuning hyperparameters.  
- Optimizing the learning outcome automatically.

This significantly reduces manual effort while maintaining high-quality model performance.

---

## 3. AutoML Approaches

There are several approaches to implementing AutoML:

### a. Bayesian Optimization
- Uses a **probabilistic model** to explore hyperparameter configurations.
- **Auto-Sklearn** is a popular example, leveraging **random forest-based sequential model-based optimization**.
- Employs **meta-learning** to find similar datasets and bootstrap optimization using prior knowledge.

### b. Recommender System Approach
- Frames AutoML as a **recommendation problem**.  
- Suggests next configurations based on similarity to previously encountered datasets.  
- Uses **probabilistic matrix factorization** to determine relationships between datasets and optimal configurations.

### c. Genetic Evolutionary Algorithms
- Uses **evolutionary techniques** to evolve pipelines and models.  
- **TPOT (Tree-based Pipeline Optimization Tool)** is a common example built around scikit-learn.

All these methods rely on multiple iterations.  
In contrast, **Oracle AutoML** employs a **non-iterative, feed-forward approach**—offering higher efficiency.

---

## 4. Oracle’s Feed-Forward AutoML Approach

**Oracle AutoML** is designed to deliver **accurate models with shorter runtimes**.  

Each stage in its pipeline makes decisions using **meta-learned proxy models**, which:
- Predict pipeline configuration performance **before** building the model.  
- Build and tune **only the best candidate pipeline**.  

This eliminates unnecessary computation and accelerates results.

### Avoiding the Cold Start Problem
Previous AutoML systems struggled with “cold start” issues—difficulty predicting pipeline performance on unseen datasets.  
Oracle AutoML solves this by:
- Leveraging **meta-learning** across a wide variety of datasets.  
- Using **pre-trained proxy models** to estimate algorithm performance accurately.

---

## 5. Zero-Code Machine Learning

Oracle AutoML can perform a **complete ML workflow without code**.  
Simply feed a dataset, and AutoML will:
- Train multiple models.
- Tune hyperparameters automatically.
- Evaluate performance across models.

This empowers **non-expert users** while also improving **data scientist productivity** by automating repetitive tasks.

### Benefits:
- Reduces manual tuning effort.  
- Decreases compute time.  
- Delivers optimal models within a given **time and budget**.

## 6. Typical AutoML Workflow

1. **Model Selection** – Choose from many candidate models.  
2. **Hyperparameter Tuning** – Adjust key parameters for better performance.  
3. **Feature Selection** – Keep only predictive features to improve speed and avoid overfitting.  
4. **Model Generalization** – Ensure strong performance on unseen data.

---

## 7. Four Core Components of Oracle AutoML

### 1️⃣ Algorithm Selection  
- Identifies algorithms that yield the **best predictive performance**.  
- Uses **meta-learning** to predict which algorithms work best for a dataset.  
- Saves time by ranking algorithm candidates without exhaustive testing.

### 2️⃣ Adaptive Sampling  
- Iteratively evaluates progressively larger samples of the data.  
- Finds the **smallest sufficient sample** to maintain model quality.  
- Detects **unbalanced datasets** early.  
- Uses **meta-learning** to estimate algorithm performance efficiently.

### 3️⃣ Feature Selection  
- Removes noisy or redundant features.  
- Uses **ranking algorithms** to score features based on importance.  
- Retains only the most predictive subset to improve accuracy and efficiency.

### 4️⃣ Model Tuning  
- Optimizes model **hyperparameters** for best performance.  
- Focuses on efficiency and scalability rather than exhaustive search.  
- Example: Tuning maximum tree depth and minimum split percentage in a decision tree.

## 8. Implementation Details in Oracle AutoML

Oracle AutoML uses the **ADS AutoML package** for training.  
Two key parameters are supported:

- **`n_jobs`** → Controls parallelism (default `-1` uses all cores).  
- **`log_level`** → Sets verbosity of output logs.

AutoML provides **pipeline summaries**, including:
- Training data information.  
- Pipeline configuration details.  
- Model trial results and metrics.

### Notes:
- Adaptive sampling is skipped if the dataset has **fewer than 1,000 rows**.  
- Use **`model_list`** to control which algorithms AutoML tests.  
- Default scoring metrics:
  - Binary Classification → `roc_auc`  
  - Multiclass Classification → `recall_macro`  
  - Regression → `neg_mean_squared_error`  

AutoML also supports a **time budget** (in seconds), stopping automatically once the budget expires and returning the best model found so far.
