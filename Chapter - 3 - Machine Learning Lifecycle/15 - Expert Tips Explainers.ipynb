## 🧠 What is Explainability?

Machine learning models can often behave like **black boxes** — they make accurate predictions, but it’s not always clear *why*.  
**Explainability** helps us understand:
- The **reasoning** behind model predictions  
- The **importance** of different features  
- The **trustworthiness** and **transparency** of the model  

In Oracle AutoMLx, Explainability is implemented through **Explainer Objects**.

The initialization sets up the necessary environment for model training, evaluation, and explainability.

### Step 2: Train the Model

As usual, we proceed with:

1. Getting an **estimator**
2. Fitting the model on the training data

```python
estimator = automlx.get_estimator(model="xgboost")
```

---

### Step 3: Obtain the Explainer Object

Once the model is trained, we can obtain an **Explainer Object** from the estimator:

```python
explainer = estimator.explainer()
```

The Explainer Object provides access to multiple **explainability methods**, including both:

* **Global Explainability** – understanding overall model behavior
* **Local Explainability** – understanding individual predictions

---

## 🧠 Explainability Methods in AutoMLx

Oracle AutoMLx provides a flexible **ML Explainer Interface** that supports multiple types of explainers depending on the dataset and model type.

### Types of Explainers

| Explainer Type       | Description                                                                      |
| :------------------- | :------------------------------------------------------------------------------- |
| **TabularExplainer** | Used for structured/tabular data (e.g., customer, transaction, or numeric data). |
| **TextExplainer**    | Used for textual datasets (e.g., reviews, comments, or documents).               |

These specialized explainers allow for tailored interpretability across different data modalities.

---

## 🔍 Using Explainability Methods

After creating the Explainer Object, you can call different methods to generate explanations:

```python
# Global explanation
explainer.global_explanation()

# Local explanation
explainer.local_explanation(sample_id=10)
```

These methods reveal how features influence predictions — either across the entire dataset (**global**) or for specific samples (**local**).

---

## 💡 Best Practices for Using Explainers

* Start by exploring **Global Explainability** to understand the overall feature importance.
* Move to **Local Explainability** for understanding specific predictions or anomalies.
* Use both **TabularExplainer** and **TextExplainer** depending on your data type.
* Always visualize results to gain intuitive insights.
