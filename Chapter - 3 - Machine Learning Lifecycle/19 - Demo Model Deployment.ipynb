
# 🧠 OCI Data Science — Model Deployment Demo

---

## 🎯 Introduction

Welcome to this hands-on session on **model deployment in Oracle Cloud Infrastructure (OCI) Data Science**.  
In this demo, we’ll walk through the steps of creating, configuring, and managing a **model deployment** inside the OCI Data Science environment.

---

## 🧭 Step 1: Accessing the Project Space

- Navigate to your OCI Data Science environment.  
- Inside your **Project Space**, click on your working project.  
  - For example, in this demo, the project is named **`test-ds`**.  
- On the **left-hand side panel**, you’ll see multiple resources:
  - Notebook Sessions  
  - Jobs  
  - Models  
  - Model Deployments  

👉 Click on **Model Deployments** to proceed.

---

## ⚙️ Step 2: Creating a Model Deployment

1. Click **Create Model Deployment**.  
2. Make sure you’re in the correct **compartment**.  
   - Example: `OCI Data Science Compartment`
3. Enter a **unique name** for your deployment (maximum 255 characters).  
   - If no name is provided, OCI automatically generates one.

💡 Example:
```

test-model-deploy

```
4. Optionally, add a **description**:  
```

"This is to test model deployment."

```

---

## 🧩 Step 3: Selecting the Model to Deploy

- Click **Select** to choose an **active model** from the **Model Catalog**.  
- Example: Select an existing active model named **`RF Classifier`**.  
- Once selected, click **Submit** to confirm.

---

## 🖥️ Step 4: Choosing VM Shape and Resource Limits

- Choose a **compute shape** for your model deployment:
- **OCPUs** → 1  
- **Memory (GB)** → 15  

- You can also specify the **number of instances** for replication.  
- Example: Select **2 instances** to allow load distribution across requests.

✅ Click **Submit** once configuration is complete.

---

## 🧾 Step 5: Enabling Logging (Optional)

Logging helps monitor the behavior and requests for your model.  
Click **Select** to configure logs:

- **Access Logs** → Capture request-level details.  
- **Predict Logs** → Capture `stdout` and `stderr` from prediction code.

---

## ⚖️ Step 6: Advanced Options — Load Balancing and Tags

In the **Advanced Options**, you can configure:
- Load Balancing Bandwidth (in Mbps)
- Resource Tags

### 🧮 Recommended Load Balancer Bandwidth Formula:

```

Payload Size (KB) × Estimated Requests per Second × 8 ÷ 1,024

```

Add **20% overhead** for traffic spikes.

**Example:**
```

Payload = 124 KB
Requests per second = 120
Recommended Bandwidth = (124 × 120 × 8) ÷ 1,024 × 1.2 = 1,152 Mbps (approx.)

```

For simplicity, you can set:
```

Load Balancer Bandwidth = 10 Mbps

```

Then click **Create**.

---

## ⏳ Step 7: Monitoring Deployment Status

- The system will take a few moments to create the deployment.  
- Once ready, the **status** will update to **Active**.  
- Click the deployment name (e.g., `test-model-deploy`) to open details.

---

## 📊 Step 8: Viewing Deployment Information

The **Model Deployment Information** page displays:

- General Information  
- Compute Details  
- Associated Model  
- Tags and Owner  
- Metrics (Success Rate, Request Count, CPU & Memory Utilization)  
- Logs (Access & Predict categories)

You can also check:
- **Work Requests** — Status should show **Succeeded (100%)**.

---

## 🌐 Step 9: Invoking the Model

You can invoke your deployed model in multiple ways:

### a. **HTTP Endpoint**
Call the model directly via REST API using the provided endpoint link.

### b. **OCI CLI Command**
Execute from terminal:
```

oci data-science model-deployment invoke ...

```

### c. **OCI Python SDK**
Programmatically call the model through the OCI Python SDK.

### d. **Java SDK**
Invoke the model using the OCI Java SDK.

---

## 🔄 Step 10: Managing the Model Deployment

You can perform the following operations on your model deployment:

| Operation | Description |
|------------|-------------|
| **Deactivate** | Temporarily disable the deployment. |
| **Reactivate** | Re-enable the deployment after deactivation. |
| **Delete** | Permanently remove the deployment. |

🧹 When deleting, confirm by entering the **exact model name**.  
Deleted deployments remain visible for **30 days** before final removal.
