## 🎯 Lesson Overview

In the previous lesson, we discussed **Global Explainability** — understanding the overall behavior of machine learning models.  
In this lesson, we will explore two additional and powerful explainability techniques:

1. **Local Explainability (LIME)** – Explains *why* a machine learning model made a specific prediction.  
2. **WhatIf Analysis** – Explores *how changes in feature values* impact model predictions.

These tools are part of **Oracle’s ADS (Accelerated Data Science)** library and are key to building transparent and interpretable models.

---

## 🧠 Local Explainability (LIME)

### What is LIME?

**LIME (Local Interpretable Model-Agnostic Explanations)** helps interpret complex models by approximating their **local behavior** with simpler, interpretable models like linear regression.

While a machine learning model may be globally complex, its **local behavior** — within a small neighborhood around a specific prediction — can often be approximated accurately using a simpler model.

---

### 🔍 How LIME Works

LIME explains **why a model made a specific prediction** by following these steps:

1. **Start with a trained machine learning model.**  
2. **Select a single observation (sample)** you want to explain.  
3. **Generate synthetic samples** around that observation (local neighborhood).  
4. **Obtain model predictions** for each generated sample.  
5. **Train a simple surrogate model** (e.g., linear model) on the generated predictions.  
6. **Use the surrogate model** to interpret how each feature influences the prediction.

---

### ⚙️ Structure of ADS LIME

ADS implements an enhanced version of LIME with **three main sections**:

#### 1. Model Section
- **Left Column:**  
  Displays model details such as:
  - True label value for the selected sample  
  - Model’s predicted value  
  - Prediction probabilities (for classification) or numeric predictions (for regression)
- **Right Column:**  
  Displays the specific **sample to explain**, showing feature names and their corresponding values.

> 📝 For text datasets, the right column displays the actual text sample instead of a table.

---

#### 2. Explainer Section
- **Left Column:**  
  Displays configuration parameters, such as:
  - Underlying local explanation algorithm (e.g., LIME)  
  - Type of surrogate model (e.g., Linear)  
  - Number of randomly generated samples (e.g., 5000)  
  - Whether continuous features are discretized  

- **Right Column:**  
  Contains a **legend** explaining how to interpret the model explanations.

---

#### 3. Explanation Section

For **classification tasks**, a local explanation can be generated for each target label.  
- In **binary classification**, explanations for one class will mirror the other.  
- In **multiclass classification**, each feature’s contribution to each class is shown separately.

##### 🧩 Feature Importance
- Visualized as a **horizontal bar chart**.
- Features at the top (larger bars) are **more influential**.  
- **Positive values (right side):** Feature increases the target prediction.  
- **Negative values (left side):** Feature decreases the target prediction.

##### 📊 Explanation Quality
Evaluates how accurately the surrogate model approximates the local behavior of the black-box model.  
Includes:
- **Sample Distance Distribution:** Highlights the locality of generated samples.
- **Evaluation Metrics:** Uses classification/regression metrics to assess surrogate model accuracy.

## 🔬 WhatIf Explainability

### Purpose
The **WhatIf Explainer** helps users understand **how changes in feature values affect model predictions**.

There are **two main techniques** in the WhatIf Explainer:

1. **Explore Sample**
2. **Explore Predictions**

---

### 🧩 Explore Sample

- Opens a **graphical interface (GUI)** showing a single observation.
- You can **change feature values manually**.  
- By clicking **Run Inference**, the model recomputes the prediction with updated values.
- Displays both **original and updated values** for easy comparison.

This is particularly useful for exploring **"what would happen if..."** scenarios for individual predictions.

---

### 📈 Explore Predictions

Allows exploration of how predictions vary across feature values.

#### 1️⃣ Single Feature (Marginal Distribution)
- When one feature (e.g., `age`) is selected as `x`, the chart shows the **relationship between that feature and the predicted outcome (y)**.  
- Example: The chart shows how the predicted risk changes as **age** increases.

#### 2️⃣ Two Features (Joint Distribution)
- When two features (e.g., `age` and `CRIM`) are passed as `x` and `y`, the plot displays:
  - One feature on the **x-axis**
  - The other feature on the **y-axis**
  - **Color gradients** to represent prediction values

This 2D visualization helps understand **feature interactions** and their combined impact on model output.

## 🧩 Summary of Key Concepts

| Concept | Description |
|:---------|:-------------|
| **Local Explainability (LIME)** | Explains *why* the model made a specific prediction. |
| **Surrogate Model** | A simple interpretable model approximating complex local behavior. |
| **Feature Importance** | Visualizes how much each feature contributes to a prediction. |
| **Explanation Quality** | Measures how well the surrogate model represents the original model locally. |
| **WhatIf Explainer** | Explores how feature changes impact predictions. |
| **Explore Sample** | Manually adjust one observation and re-run inference. |
| **Explore Predictions** | Visualize prediction trends for one or two features. |

---

## 💡 Key Takeaways

- **LIME** helps explain individual predictions by approximating the model locally.  
- **ADS LIME** provides detailed model, explainer, and explanation visualizations.  
- **WhatIf Explainer** enables interactive experimentation with feature values.  
- Together, they make model predictions more **transparent**, **trustworthy**, and **actionable**.
