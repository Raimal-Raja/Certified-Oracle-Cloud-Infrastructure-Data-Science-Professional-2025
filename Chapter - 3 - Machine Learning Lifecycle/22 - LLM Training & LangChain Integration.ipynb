# Lesson: Large Language Model Training and Language Integration  
### OCI Data Science Course

---

## 🧠 Introduction

Hello, and welcome to this lesson on **Large Language Model (LLM) Training and Language Integration**, a part of the **OCI Data Science course**.

The **OCI Data Science Jobs** feature provides a **fully managed infrastructure** that enables the training of large language models at scale.  
It supports both **full-parameter fine-tuning** and **parameter-efficient fine-tuning**, making it flexible for different model optimization approaches.

With the **Accelerated Data Science (ADS)** library, you can start training jobs by using source code directly from **GitHub** — **without making any code changes**.

---

## ⚙️ Accessing and Fine-Tuning the Model

To **fine-tune** a model, the following steps are typically involved:

1. **Access the Pre-Trained Model**  
   - You can obtain the pre-trained model from **Meta** or **Hugging Face** repositories.  
   - These pre-trained models serve as a strong foundation for domain-specific customization.

2. **Define the Training Job using ADS Python API**  
   - The training job can be easily configured with the **ADS Python SDK**.  
   - You define parameters such as model source, training data, and compute configuration.

3. **Create and Start the Job Run**  
   - The job can be created and executed through a simple **API call**.  
   - During execution, you can also **stream the job outputs** in real-time.

---

## 🔧 Behind the Scenes: Job Execution Process

Once initiated, the **training job** performs several automated steps:

- **Setup of Conda Environment** and installation of required dependencies.  
- **Fetch source code from GitHub** and checkout the specific commit.  
- **Run the training script** with arguments defining model paths, dataset location, and hyperparameters.  
- **Download the model and dataset** automatically from the specified sources.  
- **Save outputs to OCI Object Storage** once training completes successfully.

During this process, **you do not need to specify the number of nodes or GPUs** —  
ADS will **automatically configure** the resources based on the **replica and shape** you define.

---

## 💾 Saving and Managing Checkpoints

When the fine-tuning process finishes, all **model checkpoints and outputs** are automatically saved to the **OCI Object Storage bucket** that you specify in the configuration.  
These checkpoints allow you to resume training or deploy the fine-tuned model for inference later.

---

## 🤖 Integration with Generative AI and LangChain

**OCI Generative AI** provides several advanced model capabilities, such as:

- **Text Generation**  
- **Summarization**  
- **Embeddings for Semantic Search**

These can be **integrated with LangChain** using the **ADS library**.

By default, the integration uses the **same authentication method** that was set with the `ads.set_auth()` method.  
However, you can also override this by passing an `auth` keyword argument when initializing the model —  
for example, to use **Resource Principal authentication** for all OCI-related operations.
