
# Lesson: Using Data Science Operators in OCI Data Science  
### Forecasting, Anomaly Detection, and PII Detection Operators

---

## 🧭 Introduction

**Data Science Operators** are a newly introduced feature in **Oracle Cloud Infrastructure (OCI) Data Science**.  
Operators are **pre-packaged sets of libraries** designed for specific machine learning and AI tasks such as:

- Forecasting  
- Anomaly Detection  
- Detecting Personally Identifiable Information (PII)  

These operators are **low-code** and can be executed in various environments, including **OCI Data Science Notebooks**, and can even be invoked through the **OCI Command Line Interface (CLI)**.

---

## ⚙️ Types of Operators

There are **three main operator types** currently available:

1. **Forecasting Operator**  
   - Works with time-series data to predict future values.  
   - Example: Given weather data from the past five years, predict temperature for the next week.  

2. **Anomaly Detection Operator**  
   - Detects unusual or inconsistent patterns in your data.  
   - Example: Identify outliers in a dataset where certain values deviate from the norm.  

3. **PII Detection Operator**  
   - Detects and redacts **personally identifiable information** in data.  
   - Example: Mask email addresses, phone numbers, or employment history.

## 🧩 Step 2: Setting Up the Forecasting Operator

We begin by setting up the **Forecasting Operator**.  

There are several installation methods available, but here we’ll use it via the **Conda environment** in the OCI Notebook Session.

### Steps:
1. Go to **Environment Explorer** in your OCI Data Science Notebook.
2. Locate the **AI Forecasting Operator** pre-created environment.
3. Review the libraries included within this environment.  
   These libraries are used to train models and generate forecasts.
4. Copy the **installation command** from Environment Explorer.
5. Open a **terminal** in your notebook and paste the command to install.

In many cases, Conda environments for **Forecasting** and **Anomaly Detection** are **pre-created**, so you can directly activate them without reinstallation.

---

## 🧱 Step 3: Initializing the Operator

Navigate to the **Quick Start section** in the documentation and locate the **initialize command**.  
Copy this command into your terminal and execute it.

When executed, it creates a new directory — for example, `my-forecast`.

### Directory Structure:
Inside `my-forecast`, several YAML configuration files are generated automatically.  
These YAML files define settings for running the operator in different environments (local or cloud).

---

## 🧾 Step 4: Configuring the YAML File

1. Open the `my-forecast` directory.  
2. Edit the `forecast.yml` configuration file.  
3. You will see a few pre-filled parameters.  
4. Copy the updated configuration content from the documentation and paste it here.  

This YAML file defines:
- Input dataset location  
- Forecasting horizon  
- Model parameters  
- Output directory  

Once the file is configured, **save it**.

---

## 🧮 Step 5: Activating the Environment and Running the Operator

1. **Activate the Conda environment** for the forecasting operator.  
   ```bash
   conda activate ai-forecasting
````

2. **Navigate** to the `my-forecast` directory:

   ```bash
   cd my-forecast
   ```

3. **Run the operator** using the provided CLI command:

   ```bash
   ads opctl run -f forecast.yml
   ```

The operator will start execution.
It may take a few minutes to process the data, train the model, and generate results.

---

## 📊 Step 6: Viewing the Results

Once the operator finishes execution, it creates an output directory, typically named **`results`**.

### Contents:

* **forecast.csv** – Contains forecasted data points.
* **report.html** – A complete HTML summary report.

When you open the **report.html** file, you’ll see:

* The first and last 10 rows of the dataset.
* Statistical summaries.
* Forecasted data overlaid with historical trends.

---

## 🪄 Step 7: Understanding the Workflow

The process to use an operator follows a simple, repeatable workflow:

1. **Install** the required Conda environment.
2. **Initialize** the operator to create configuration files.
3. **Activate** the Conda environment.
4. **Edit** the YAML configuration file (e.g., `forecast.yml`).
5. **Run** the operator.
6. **Review** the results.

For advanced customization, consult the **YAML Schema section** in the documentation.

---

## ⚡ Step 8: Using Other Operators

### 🔍 Anomaly Detection Operator

* Install or activate the corresponding Conda environment.
* Initialize the operator.
* Edit the YAML file for configuration.
* Run the operator using the CLI command.

This identifies **anomalous data points** that deviate from expected patterns.

---

### 🔐 PII Detection Operator

* Install or activate the **PII Conda environment**.
* Initialize the operator using the command:

  ```bash
  ads opctl init -o pii
  ```
* Configure the YAML file.
* Run the operator using:

  ```bash
  ads opctl run -f pii.yml
  ```

This operator identifies and redacts sensitive information like **emails, phone numbers, and employment history**.

---

## ✅ Summary

| Operator              | Purpose                                      | Typical Use Case                          |
| --------------------- | -------------------------------------------- | ----------------------------------------- |
| **Forecasting**       | Predict future values using time-series data | Temperature prediction, sales forecasting |
| **Anomaly Detection** | Detect irregular or unexpected data points   | Fraud detection, data quality validation  |
| **PII Detection**     | Identify and redact sensitive personal data  | Data anonymization, compliance tasks      |

---

## 🏁 Conclusion

In this lesson, we explored how to:

* Install and configure **OCI Data Science Operators**
* Use **Forecasting**, **Anomaly Detection**, and **PII Detection** operators
* Understand YAML-based configuration and Conda environment setup
* Run and review operator outputs in **OCI Data Science**

These operators offer a **low-code**, **scalable**, and **efficient** way to perform key data science tasks such as forecasting, anomaly detection, and PII redaction — directly within the OCI ecosystem.

---

**End of Lesson: Data Science Operators in OCI Data Science**

```
