## 2. What is Model Training?

Model training builds the **best mathematical representation** of relationships:

- Among **features and a target label** (in *supervised learning*).  
- Among **features themselves** (in *unsupervised learning*).

The process creates a **model artifact** that captures these relationships or patterns.  
It determines the **best algorithm** for training, taking into account multiple dimensions of the data.

---

## 3. The Model Training Process

A typical **model training workflow** involves several core functions:

### a. Score Function
- Indicates how well the model fits the data.  
- Could represent:
  - **Error function** (to minimize)
  - **Maximum likelihood function** (to maximize)

### b. Loss Function
- Also known as the **cost function**.  
- Compares predictions made by the model to the **true target values**.  
- Represents the difference as a single number — the **loss score**.

**Illustration:**
- Green dots → true data points  
- Black line → model predictions  
- Red arrows → the loss (difference between prediction and actual value)

### c. Update Function
- Updates model parameters **iteratively** from one step to the next to minimize loss and improve accuracy.
