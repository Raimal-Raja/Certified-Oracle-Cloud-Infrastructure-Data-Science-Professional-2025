## 2. Training an AML Model on OCI

You can **easily train a simple AML model** using **Jobs** in the **OCI Data Science Service**.

### Workflow Overview:
1. Use **GitHub** to host the **source code**.  
2. Use **OCI Object Storage** to store **training results** and **artifacts**.  
3. With **ADS (Accelerated Data Science)**:
   - Create a **Job** to define compute and resource configurations.  
   - Create a **Job Run** to execute the training script.  
   - Save the training outputs automatically.

You can set up these configurations using either:
- **Python code**, or  
- **YAML configuration files**.

## 3. Distributed Training on OCI

**Distributed training** enables horizontal scaling — parallelizing training tasks for **large datasets** or **compute-intensive workloads**.  
It helps reduce training time without compromising accuracy.

Using **OCI Data Science Service with ADS**, you can perform distributed training with popular frameworks like:

- **Dask**  
- **PyTorch Distributed**  
- **Horovod**  
- **TensorFlow Distributed**

### Key Benefits:
- Scales horizontally across compute nodes.  
- Reduces model training time significantly.  
- Maintains accuracy and reproducibility.  

To practice distributed training:
- Review the **OCI documentation** for step-by-step implementation details.  
- Implement use cases on your own using **Docker** or **GitHub** setups.  
- Share your projects on **GitHub** and in the **Oracle University (OU) community** for collaboration.

## 4. AutoML and AutoMLx Package

In this course, we also explored **AutoML** concepts.  
Now, let’s look at the **AutoMLx** package — available in the **OCI Data Science Service** as part of the **AutoMLx Conda environment**.

### Oracle AutoMLx Solution:
- Provides a **pipeline** that automatically:
  - Finds the optimal model for a prediction task.  
  - Tunes hyperparameters automatically.  
  - Uses a given **training dataset** and **target variable**.  

You can choose between:
- **Local** engine  
- **Task-parallel** engine  
(both managed using the **INIT()** function)
