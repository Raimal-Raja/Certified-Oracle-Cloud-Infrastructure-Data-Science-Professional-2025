
This ensures all subsequent interactions — model creation, storage, and deployment — are authorized through OCI’s identity framework.

---

## 🧠 Step 3: Create Embeddings and Model

* **Embeddings Creation:**
  We use the `GenerativeAIEmbeddings` class to create **vector representations** of textual data. These embeddings capture the semantic meaning of the content.

* **Model Creation:**
  We instantiate a model using the **GenerativeAI** class, which represents the **LLM (Large Language Model)** that will later perform the retrieval-augmented reasoning.

---

## 📄 Step 4: Load and Process Documents

1. **Create a Text Loader** — used to read a document into memory.
2. **Load the Document** — import text content from a file or a given text input.
3. **Split the Document** — break the document into smaller, semantically meaningful chunks for better retrieval efficiency.
4. **Create a Vector Data Store** — using the documents and their corresponding embeddings.

This vector database allows for **semantic retrieval** based on meaning, not just keywords.

---

## 🔍 Step 5: Create a Retriever

We create a **retriever** using our vector database.
This retriever enables **contextual information lookup** during question-answering tasks by finding the most relevant text segments from our document set.

---

## 🔗 Step 6: Build the Retrieval Chain

We now create a **RetrievalQA chain** — a LangChain construct that connects the retriever to the LLM.

* The **LLM** performs the reasoning.
* The **retriever** provides context from the stored documents.

By combining the two, the model can answer user queries with information derived directly from your own data.

---

## 🗂️ Step 7: Prepare Model Artifacts

Next, we create a **temporary directory** to store model artifacts, and then we use the **`ChainDeployment` class** to prepare the model for deployment.

```python
model.prepare()
```

This method packages the model files, configuration, and metadata required for OCI deployment.

---

## 🧪 Step 8: Verify the Model

Before deploying, we verify that the model is functioning correctly.
We send a test query to ensure that the prepared model returns coherent responses.

**Example Verification Prompt:**

> “What is AI Foundations course?”

✅ The model responds accurately, confirming that the deployment artifacts are valid and operational.

---

## 💾 Step 9: Save and Deploy the Model

* **Save Step:**
  We use the `.save()` method to store the model into the **OCI Model Catalog**.

* **Deploy Step:**
  The `.deploy()` method publishes the model as a **live endpoint** on OCI Data Science.

Deployment typically takes **10–15 minutes**, after which the model becomes available for inference.

---

## 🧭 Step 10: Test the Deployed Model

Once the model is deployed, we use the `.predict()` method to query it.

**Example Query:**

> “Who are the instructors for AI Foundation?”

🧩 The model retrieves this **specific information** from the document we provided earlier and responds with:

> “The instructors for the Oracle Cloud Infrastructure AI Foundation course are Hemant Gahankari, Himansha Raj, and Nick Commisso.”

This demonstrates that the model successfully performs **retrieval-augmented question answering** from the ingested content.

---
