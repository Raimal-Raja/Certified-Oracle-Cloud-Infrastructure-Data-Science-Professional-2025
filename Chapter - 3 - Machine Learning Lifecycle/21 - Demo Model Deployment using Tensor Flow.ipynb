## 🎓 Introduction

Welcome to this demo: **Model Deployment using TensorFlow Model Class in OCI Data Science**.  
Here, we will explore how to register and deploy your model through the **Accelerated Data Science (ADS)** library.

The **ADS library** provides framework-specific classes that simplify model registration and deployment to production with just a few lines of code.

---

## ⚙️ Step 1: Creating the Model Serialization Object

The first step is to **create a model serialization object**.  
This object wraps your trained model and offers several built-in methods that assist with deploying it.

Different frameworks use different model classes:

- `PyTorchModel` → for PyTorch models  
- `TensorFlowModel` → for TensorFlow models  
- `SklearnModel` → for scikit-learn models  

In this demo, we will showcase deployment using a **TensorFlow model class**.

```python
from ads.model.framework.tensorflow_model import TensorFlowModel
````

---

## 🧩 Step 2: Preparing Model Artifacts

After creating the model serialization object, use the `.prepare()` method to generate the **model artifacts**.
The ADS library automatically creates a **score.py** file tailored to your model type.

You may optionally customize this file, but in most cases, it works as-is.

The `prepare()` method can also store:

* Model metadata
* Training code
* Input/output schema
* Other relevant configurations

If you modify the `score.py` file, run the `.verify()` method to confirm the `load_model()` and `predict()` functions are functioning correctly before deployment.

```python
tf_model.prepare(
    inference_conda_env="tensorflow28_p37_cpu",
    force_overwrite=True
)
```

---

## 🧠 Step 3: Verifying the Model

The `.verify()` method helps test your model locally before deployment.
It confirms that `score.py` loads and predicts correctly without deploying to production.

```python
tf_model.verify()
```

This saves debugging time and avoids unnecessary redeployments.

---

## 💾 Step 4: Saving the Model to the Model Catalog

Once verification is complete, use the `.save()` method to register your model into the **OCI Data Science Model Catalog**.
This catalog maintains versioned and managed records of your models.

```python
tf_model.save(display_name="Demo_FashionMNIST_TF_Model")
```

Upon successful registration, you’ll receive a unique **Model OCID** for tracking and reference.

---

## 🚀 Step 5: Deploying the Model

The `.deploy()` method creates an HTTPS endpoint for inference, complete with a **load balancer** and **virtual machine instances**.

You can specify:

* Display name
* Instance type and count
* Bandwidth limits
* Logging configurations

```python
deployment = tf_model.deploy(display_name="Demo_TF_Model_Deployment")
```

Once deployment is active, the model is ready to accept real-time prediction requests.

---

## 🧾 Step 6: Making Predictions

With your model deployed, you can use the `.predict()` method to send data to the model endpoint and receive predictions.

```python
data_sample = x_test[:5]
predictions = tf_model.predict(data_sample)
print(predictions)
```

This performs inference against the live model through the deployed HTTPS endpoint.

---

## 🧱 Libraries Used in This Demo

Below are the major Python modules used throughout the process:

| Library               | Purpose                               |
| --------------------- | ------------------------------------- |
| `ads`                 | Interaction with OCI Data Science     |
| `logging`             | Manage logging output                 |
| `os`                  | File and path operations              |
| `pandas`              | Data manipulation                     |
| `tempfile`            | Temporary directories and files       |
| `tensorflow`          | Deep learning framework               |
| `tensorflow_datasets` | Access to datasets like Fashion-MNIST |
| `matplotlib`          | Visualization of datasets             |
| `shutil`              | Clean-up of directories               |
| `warnings`            | Suppress unnecessary warnings         |

---

## 🧮 About the Fashion-MNIST Dataset

The **Fashion-MNIST** dataset contains:

* 60,000 training samples
* 10,000 testing samples
* Each image is **28×28 grayscale**
* 10 output classes representing clothing categories

Below is a code example to load and visualize sample data:

```python
import tensorflow as tf
import matplotlib.pyplot as plt

(train_images, train_labels), (test_images, test_labels) = tf.keras.datasets.fashion_mnist.load_data()

plt.figure(figsize=(6,6))
for i in range(9):
    plt.subplot(3,3,i+1)
    plt.imshow(train_images[i], cmap='gray')
    plt.axis('off')
plt.show()
```

---

## 🧱 Step 7: Building and Training the TensorFlow Model

We define a simple **Sequential Neural Network** for classification:

```python
from tensorflow.keras import models, layers

model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.2),
    layers.Dense(10, activation='softmax')
])

model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

model.fit(train_images[:10000], train_labels[:10000], epochs=5)
```

Output (example):

```
Loss: 0.7899 | Accuracy: 0.7235
```

---

## 📦 Step 8: Preparing Artifacts for Deployment

When `.prepare()` is executed, the following files are generated:

| File                 | Description                                   |
| -------------------- | --------------------------------------------- |
| `input_schema.json`  | Defines input data structure                  |
| `model.h5`           | Serialized model file                         |
| `output_schema.json` | Defines output data structure                 |
| `runtime.yaml`       | Defines runtime and environment configuration |
| `score.py`           | Contains `load_model()` and `predict()` logic |

You can view deployment progress using:

```python
tf_model.summary_status()
```

This shows each step — *initiate, prepare, verify, save, deploy, predict* — with its status and details.

---

## 🧾 Step 9: Verifying and Viewing Metadata

Metadata attributes in the TensorFlow model object include:

* `runtime` → Deployment environment info
* `metadata_input` → Input feature schema
* `metadata_custom` → Custom metadata key-value pairs
* `metadata_provenance` → Training data and code provenance
* `metadata_taxonomy` → Classification and parameters

These ensure full transparency and reproducibility in your ML workflow.

---

## 🧰 Step 10: Managing and Monitoring Deployment

Once active, you can monitor the deployment in the OCI console:

* View CPU/memory/network utilization
* Check access and prediction logs
* View model configuration and compute shape
* Modify instance count or description with zero downtime

Use the **OCI Logging Service** and **Monitoring Dashboard** for real-time visibility into model performance.

---

## 🧹 Step 11: Cleaning Up Resources

Once the demo is complete, it’s best practice to delete the deployment and model to free resources.

```python
# Delete deployment first
tf_model.delete_deployment()

# Delete model from the catalog
tf_model.delete()

# Clean up local artifacts
import shutil
shutil.rmtree(model_artifact_dir, ignore_errors=True)
```

---

## ✅ Summary

Let’s recap the end-to-end process of deploying a TensorFlow model using the ADS library:

1. **Create** the TensorFlow model object
2. **Prepare** the model artifacts
3. **Verify** model correctness
4. **Save** it to the Model Catalog
5. **Deploy** the model as an HTTPS endpoint
6. **Predict** using live inference
7. **Monitor** and **clean up** resources

With these steps, you can take any trained TensorFlow model and push it to production on **OCI Data Science** quickly and efficiently.

---
