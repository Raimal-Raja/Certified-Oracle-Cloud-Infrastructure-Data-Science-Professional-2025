## **2. Understanding Hyperparameters**

**Definition:**
- **Hyperparameters** are model configuration settings that control the learning process.
- They are **not** learned from the training data directly.
- Instead, they guide how the model learns from data.

**Examples:**
- Learning rate
- Maximum depth of a decision tree
- Number of estimators in ensemble models
- Regularization parameters

---

## **3. What is Hyperparameter Tuning?**

**Hyperparameter tuning** is the process of:
1. Searching for the **optimal set of hyperparameter values**.
2. Evaluating model performance for each combination.
3. Selecting the configuration that yields the **best model results**.

This process is **iterative** — multiple model training runs are performed with different hyperparameter combinations until the best configuration is found.

---

## **4. Role of ADSTuner**

**ADSTuner** is Oracle’s hyperparameter tuning utility included in the ADS SDK.

### **Key Features:**
- Supports **multiple hyperparameter search strategies**.
- Works with **any ML library** (even those lacking built-in tuning features).
- Allows **user-defined search spaces and strategies**.
- Generates a **tuning report** with:
  - All trials
  - Best-performing configurations
  - Performance statistics

---

## **5. Instantiating ADSTuner**

To use ADSTuner, you first need a reference to the model you want to tune.

### **Basic Initialization Example:**
```python
from ads.model import ADSTuner

tuner = ADSTuner(
    model=my_model,           # reference to your model
    cv=5,                     # number of cross-validation folds (optional)
    strategy="perfunctory"    # search strategy (optional)
)
````

* `model`: The model instance (e.g., LogisticRegression, XGBoostClassifier, etc.)
* `cv`: Number of cross-validation folds to use
* `strategy`: The hyperparameter search approach

---

## **6. Search Strategies**

ADSTuner supports several **search space strategies**, controlling how many parameters are tuned and how widely they are searched.

### **a. Perfunctory Search Space**

* Focuses on a **small number of key hyperparameters**.
* Designed to be **computationally efficient**.
* Best used **early in experimentation** to:

  * Quickly assess model class performance.
  * Reduce compute cost and tuning time.

### **b. Detailed Search Space**

* Explores a **broader search space**.
* Tunes **more hyperparameters** with a wider range of values.
* Used **after identifying** the most suitable model class for your problem.
* Ideal for **fine-tuning** final model performance.

---

## **7. Custom Search Spaces**

When you have prior experience or domain knowledge, you can define a **custom search space**.

### **Example:**

```python
custom_space = {
    "learning_rate": [0.01, 0.05, 0.1],
    "max_depth": [3, 5, 7, 9],
    "n_estimators": [100, 200, 300]
}

tuner = ADSTuner(model=my_model, strategy=custom_space)
```

You can modify this search space by:

* **Adding** new hyperparameters.
* **Removing** unnecessary parameters.
* **Adjusting** value ranges for numeric (non-categorical) parameters.

---

## **8. Executing the Tuning Process**

Once the tuner is set up, run the tuning process with the `tune()` method:

```python
tuner.tune(
    X_train,           # feature dataset
    y_train,           # target labels
    exit_criterion={"MAX_ITERATIONS": 20}
)
```

* **`X_train`, `y_train`** — the input training data.
* **`exit_criterion`** — stopping conditions for the tuner.

---

## **9. Exit Criteria (Stopping Conditions)**

ADSTuner continues training until **one or more stopping criteria** are met.
These criteria control how long the tuning process runs.

### **Common Exit Criteria:**

* `MAX_ITERATIONS`: Maximum number of tuning trials.
* `MAX_TIME`: Maximum time duration for tuning.
* `MAX_SCORE`: Stop when desired performance is achieved.

**Example:**

```python
exit_criterion = {"MAX_ITERATIONS": 50, "MAX_TIME": 3600}
```

This will stop the tuner after 50 iterations or 1 hour — whichever comes first.

---

## **10. Cross-Validation Support**

ADSTuner integrates **cross-validation (CV)** to ensure model generalization.

* Each hyperparameter combination is validated using CV.
* Prevents overfitting and improves reliability.
* Number of folds (`cv`) can be customized (default = 5).

**Example:**

```python
tuner = ADSTuner(model=my_model, cv=10)
```

---

## **11. Tuning Report**

After tuning, ADSTuner generates a **comprehensive report** summarizing:

* All hyperparameter combinations tested.
* Corresponding performance metrics.
* The **best-performing configuration**.
* Visualizations of tuning progress (if enabled).

**Example to view the report:**

```python
tuner.report
```

---

## **12. Summary**

In this lesson, we explored **hyperparameter tuning using ADSTuner**.
Here are the key takeaways:

| Concept              | Description                                                            |
| -------------------- | ---------------------------------------------------------------------- |
| **Hyperparameters**  | Control model learning behavior, not learned from data                 |
| **ADSTuner**         | Oracle ADS tool for automated hyperparameter tuning                    |
| **Search Spaces**    | Perfunctory (small/fast), Detailed (large/deep), Custom (user-defined) |
| **Cross-Validation** | Ensures model generalization during tuning                             |
| **Exit Criteria**    | Defines when tuning stops (max iterations, time, or score)             |

**Outcome:**
ADSTuner enables efficient, flexible, and automated hyperparameter tuning — reducing manual experimentation while improving model performance.
