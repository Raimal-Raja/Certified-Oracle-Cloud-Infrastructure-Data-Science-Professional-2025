## **2. Understanding Model Evaluation**

Model evaluation is a **critical step** in the **machine learning lifecycle**, which comes **after model training**.

### **Purpose:**
Models are only as valuable as the quality of their predictions. Evaluation helps determine:
- How well the model performs on unseen data.
- Whether the predictions are reliable and generalizable.
- Which model performs best among alternatives.

### **Typical Process:**
1. Split the dataset into **training**, **validation**, and **test** sets.
2. Train the model on the training data.
3. Predict on the validation/test set.
4. Compare predicted values against actual labels using evaluation metrics.

These metrics transform model outputs into interpretable and standardized **scores and charts**.

---

## **3. Benefits of Model Evaluation**

| **Benefit** | **Description** |
|--------------|----------------|
| **Benchmarking** | Compare performance across models using standard metrics and confusion matrices. |
| **Identifying Pitfalls** | Detect issues like high accuracy but low precision, or overfitting. |
| **Understanding Trade-offs** | Evaluate how models behave under different conditions (e.g., Model A works better in clear weather). |

## **6. Binary Classification Evaluation**

Binary classification deals with **two classes** — such as “Yes/No” or “True/False”.

### **Common Metrics:**
- Accuracy  
- Precision  
- Recall  
- F1-Score  
- ROC AUC  
- Confusion Matrix

### **Example Using ADS:**
```python
from ads.model import ADSEvaluator, ADSModel
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# Convert fitted estimators to ADS Models
lr_model = ADSModel.from_estimator(LogisticRegression().fit(X_train, y_train))
rf_model = ADSModel.from_estimator(RandomForestClassifier().fit(X_train, y_train))

# Initialize Evaluator
evaluator = ADSEvaluator(
    models=[lr_model, rf_model],
    X=X_test,
    y=y_test,
    mode='classification'
)

# View metrics
evaluator.metrics
````

### **Output Example:**

| Model                  | Precision | Recall | F1-Score |
| ---------------------- | --------- | ------ | -------- |
| LogisticRegression     | 0.9024    | 0.8742 | 0.8870   |
| RandomForestClassifier | 0.8814    | 0.8920 | 0.8865   |

---

### **Displaying Charts**

```python
evaluator.show_in_notebook(perfect=True)
```

If `perfect=True`, ADS displays a **perfect classifier** for comparison in Lift and Gain charts.

### **Adding a Custom Metric**

```python
from sklearn.metrics import fbeta_score

evaluator.add_metrics({
    "F2_Score": lambda y_true, y_pred: fbeta_score(y_true, y_pred, beta=2)
})
```

---

## **7. Binary Classification Charts**

ADS automatically generates several charts for visual interpretation:

* **Lift and Gain Charts**
* **Precision-Recall Curve**
* **ROC Curve**
* **Normalized Confusion Matrix**

These help visualize model strength and trade-offs.

---

## **8. Multiclass Classification Evaluation**

For multiclass problems (e.g., predicting "Red", "Blue", or "Green"):

### **Common Metrics:**

* Hamming Loss
* Macro, Micro, and Weighted Precision
* Recall and F1-Score
* ROC/AUC for multiple classes

### **Example:**

```python
evaluator_multi = ADSEvaluator(
    models=[lr_model, rf_model],
    X=X_test,
    y=y_test,
    mode='multiclass',
    classes=[0, 1, 2]
)

evaluator_multi.metrics
```

### **Typical Output:**

| Metric             | LogisticRegression | RandomForestClassifier |
| ------------------ | ------------------ | ---------------------- |
| Macro F1           | 0.812              | 0.788                  |
| Micro F1           | 0.841              | 0.835                  |
| Weighted Precision | 0.845              | 0.830                  |

### **Charts Produced:**

* **Multiclass ROC Curve**
* **Precision by Label**
* **F1 by Label**
* **Multiclass PR Curve**

```python
evaluator_multi.show_in_notebook()
```

---

## **9. Regression Evaluation**

Regression models predict **continuous numerical values**, such as prices or lengths.

### **Common Metrics:**

| **Metric**                            | **Description**                                                |
| ------------------------------------- | -------------------------------------------------------------- |
| **R² (Coefficient of Determination)** | Measures goodness of fit                                       |
| **Explained Variance Score**          | How much variance is explained by the model                    |
| **Mean Squared Error (MSE)**          | Average squared difference between actual and predicted values |
| **Mean Absolute Error (MAE)**         | Average absolute difference                                    |
| **Root Mean Squared Error (RMSE)**    | Square root of MSE                                             |
| **Mean Residuals**                    | Average residual error                                         |

### **Example Using ADS:**

```python
evaluator_reg = ADSEvaluator(
    models=[reg_model],
    X=X_test,
    y=y_test,
    mode='regression'
)

evaluator_reg.metrics
evaluator_reg.show_in_notebook()
```

---

## **10. Regression Charts**

ADS provides visualizations for regression diagnostics:

| **Chart**                  | **Description**                                |
| -------------------------- | ---------------------------------------------- |
| **Observed vs Predicted**  | Compares actual vs predicted values            |
| **Residuals QQ Plot**      | Checks if residuals follow normal distribution |
| **Residuals vs Predicted** | Detects non-linearity and bias                 |
| **Residuals vs Observed**  | Shows error trends with actual values          |

A good model’s QQ plot points will align closely along a straight line.

---

## **11. Integration with Open Source Tools**

ADS works seamlessly with **open-source frameworks** like **scikit-learn**, allowing you to:

* Use familiar functions and metrics.
* Combine ADS visualization with scikit-learn models.
* Extend ADS evaluators with custom scoring functions.

---

## **12. Summary**

| **Concept**          | **Key Takeaway**                                                      |
| -------------------- | --------------------------------------------------------------------- |
| **Model Evaluation** | Determines the accuracy and quality of predictions                    |
| **ADS Evaluators**   | Simplify metric computation and visualization                         |
| **Evaluation Types** | Binary, Multiclass, Regression                                        |
| **Custom Metrics**   | Add and visualize new metrics using ADS                               |
| **Charts**           | Gain insight into performance using Lift, ROC, PR, and Residual plots |

---
