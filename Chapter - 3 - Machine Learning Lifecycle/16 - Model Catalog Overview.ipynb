## 🎯 Lesson Overview

In this lesson, we’ll explore the **Model Catalog** in Oracle Cloud Infrastructure (OCI) Data Science.  

The Model Catalog provides a **centralized and immutable repository** for managing machine learning (ML) models throughout their lifecycle.  
You’ll learn how it helps maintain **model provenance**, ensure **traceability**, and support **collaboration** across teams.

By the end of this lesson, you will understand:
- The **purpose and structure** of the Model Catalog  
- The **components** of a model artifact  
- How **immutability** ensures model reliability  
- How to prepare **model metadata**, **schemas**, and **taxonomy**

---

## 🧠 What is the Model Catalog?

The **Model Catalog** provides a method to **track**, **store**, and **manage** ML models in an immutable way.  
It enables organizations to maintain **provenance** — a detailed history of how and where a model was created, trained, and deployed.

### Key Benefits:
- **Traceability:** Track every version of a model throughout its lifecycle  
- **Immutability:** Prevent unintentional changes to deployed models  
- **Reproducibility:** Easily recreate results using stored model artifacts  
- **Collaboration:** Share models across data science teams  

---

## 🧩 Components of a Model Entry

Each entry in the Model Catalog contains multiple components that collectively define the model and its metadata.

| Component | Description |
|:------------|:-------------|
| **Model Artifact** | A ZIP file containing the trained model and related files |
| **score.py** | Python script that loads the model and defines the inference logic |
| **runtime.yaml** | Specifies the runtime environment (Conda, Python version, etc.) |
| **Metadata** | Includes provenance information, Git details, and training environment |
| **Push Script/Notebook** | The script or notebook used to upload the model to the catalog |

---

## 🔒 Immutability and Versioning

Models stored in the catalog are **immutable** by design.  
To make any modifications, you must **create a new model version** rather than editing an existing one.

**Benefits of immutability:**
- Prevents unauthorized or accidental changes  
- Ensures that every deployed model can be traced back to its exact artifact  
- Enhances compliance and audit readiness  

---

## 📦 Model Artifact Details

A **model artifact** is a compressed ZIP file that includes all the files needed to reproduce and deploy a model.

### Included Files:

| File | Purpose |
|:------|:----------|
| **score.py** | Contains logic to load serialized model objects and define an inference endpoint |
| **runtime.yaml** | Defines the Conda environment and dependencies for deployment |
| **validate.py** | Optional file for running introspection tests before saving to the catalog |
| **requirements.txt** | Lists third-party Python dependencies |
| **README.md** | Step-by-step instructions to prepare and save the model artifact |

---

## ⚙️ Model Artifact Structure

When packaging your model artifact, ensure that all required files are organized correctly.

```

model_artifact/
│
├── score.py
├── runtime.yaml
├── validate.py
├── requirements.txt
└── README.md

````

> **Important:**  
> Any code used for inference must be zipped at the same level as `score.py`.  
> Files located above that level will be ignored and may cause deployment failures.

---

## 🧾 score.py File

`score.py` defines how your model loads and predicts.  

It typically includes:
- A **load_model()** function — loads the serialized model into memory  
- A **predict()** function — defines how predictions are made  
- Optional **helper functions** — for custom preprocessing or transformations  

You can customize function parameters as needed for your model’s requirements.

---

## 🧮 runtime.yaml File

This file defines the **runtime environment** that the model will use during deployment.

| Field | Description |
|:--------|:-------------|
| **artifact_version** | Version of the artifact format (automatically extracted by ADS) |
| **inference_environment_slug** | Conda environment slug for scoring |
| **inference_environment_type** | Either `dataScience` or `published` |
| **inference_environment_path** | Object Storage path to the Conda environment (e.g., `bucket@namespace/path`) |
| **python_version** | Python version for deployment (default: 3.6; supported: 3.6 or 3.7) |

> Typically, the **inference environment** matches the **training environment**, but they can differ if needed.

---

## 🧠 Model Metadata and Documentation

The Model Catalog supports multiple documentation types that enrich your model metadata and improve reproducibility.

### Types of Model Documentation

| Type | Purpose |
|:------|:----------|
| **Input/Output Schema** | Describes input features and output structure for model predictions |
| **Model Provenance** | Captures Git info, training environment, and data lineage |
| **Introspection Tests** | Validates operational health and detects common errors |
| **Model Taxonomy** | Describes the model’s use case, framework, algorithm, and custom metadata |

---

## 📋 Input and Output Schema

Schemas define the structure of data used to make predictions.  
They act as **contracts** between the model and its consumers.

### Example Input Schema
```json
{
  "features": [
    {"name": "age", "type": "integer"},
    {"name": "income", "type": "float"},
    {"name": "education", "type": "string"}
  ]
}
````

### Example Output Schema

```json
{
  "prediction": {"type": "string"}
}
```

> These schemas are primarily for documentation purposes in the current release.

## 🧬 Model Provenance

Model provenance ensures that every model can be traced to:

* The **training data and features**
* The **source code** and **Git commit**
* The **compute resources** used
* The **training environment**

If you are working inside a Git repository, ADS automatically captures Git-related information for provenance metadata.

---

## 🧪 Model Introspection Tests

Introspection tests are optional but highly recommended.
They check for:

* Model integrity
* Compatibility
* Correctness of serialization and loading

After running, results are saved in a local file named:

```
test_json_output.json
```

These results can be uploaded to the catalog as part of model metadata.

---

## 🧾 Model Taxonomy

Model taxonomy provides structured documentation that describes the model and its technical context.

| Field                     | Description                                                       |
| :------------------------ | :---------------------------------------------------------------- |
| **Use Case Type**         | ML use case (e.g., Binary Classification, Regression, Clustering) |
| **Framework**             | ML framework (e.g., scikit-learn, TensorFlow, PyTorch)            |
| **Framework Version**     | Framework version (e.g., PyTorch 1.9)                             |
| **Algorithm**             | Algorithm or class used (e.g., CART, Logistic Regression)         |
| **Hyperparameters**       | Model hyperparameters in JSON format                              |
| **Artifact Test Results** | JSON output of introspection tests                                |
| **Custom Metadata**       | User-defined metadata fields (key, value, category, description)  |

> **Note:** The combined size of defined and custom metadata cannot exceed **32 KB**.

---

## 🧭 Model Catalog Policies and Access

Just like other OCI resources, access to the **Model Catalog** is governed by **OCI IAM policies**.

You must define policies for:

* **Model catalog management**
* **Model invocation**
* **Cross-service access** (e.g., Object Storage access for artifacts)

These policies ensure that only authorized users can push, pull, and deploy models.
