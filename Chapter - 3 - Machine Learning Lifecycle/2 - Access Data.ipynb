## 1. Oracle Object Storage

- Load data from Object Storage into a dataframe using the **ADS library**.  
- Replace placeholders with your actual bucket name and file path.

Example:
```python
ads.set_auth(resource_principal=True)
df = pd.read_csv("oci://<bucket-name>@<namespace>/<filename>.csv")
````

* **Authentication Options:**

  * **API Key**
  * **Resource Principal** (preferred for serverless functions)
* The `set_auth` module toggles between key pair or resource principal authentication.

---

## 2. Local Storage

* Use **pandas** to read data directly from local files.

```python
df = pd.read_csv("/path/to/local/file.csv")
```

---

## 3. Oracle Autonomous Databases (ADB)

* OCI supports both Autonomous Database services.
* Use the **ADS optimized read function**:

```python
df = ads.read_sql("SELECT * FROM my_table", connection_parameters)
```

* It is up to **15× faster** than pandas’ `read_sql()` since it bypasses ORM layers.
* If using a **wallet file**, define:

  * Connection parameters
  * Wallet location

Example:

```python
connection_parameters = {
    "service_name": "mydb_high",
    "wallet_location": "/path/to/wallet"
}
```

* For **connections without a wallet**, specify `hostname` and `port`:

```python
connection_parameters = {
    "host": "mydb.oci.oraclecloud.com",
    "port": 1521,
    "service_name": "mydb_high"
}
```

> ⚙️ Requires ADS version **2.5.6 or later**.

* Use **bind variables** to prevent **SQL injection attacks**.
* Performance can depend on network latency and SQL efficiency.

---

## 4. MySQL Databases

* Follow similar steps as ADB, but set the engine as `"mysql"`.

```python
df = ads.read_sql("SELECT * FROM my_table", engine="mysql")
```

* To save data back to MySQL:

```python
ads.to_sql(df, "my_table", engine="mysql")
```

> ✅ Available in ADS version **2.5.6+**.

---

## 5. Amazon S3

* Access **public or private** files via pandas.
* For private files, pass credentials in the `storage_options` dictionary:

```python
df = pd.read_csv("s3://bucket/file.csv", storage_options={"key": "...", "secret": "..."})
```

---

## 6. HTTP/HTTPS Endpoints

* Access data directly through a URL using pandas:

```python
df = pd.read_csv("https://example.com/data.csv")
```

---

## 7. DatasetBrowser

* **ADS** provides access to pre-defined datasets from:

  * Seaborn
  * scikit-learn
  * GitHub
  * And more

List available datasets:

```python
ads.dataset.list()
```

Open a specific dataset:

```python
ads.dataset.open("iris", source="seaborn")
```

---

## 8. PyArrow and OCI File Systems

* OCI FS (File System) library provides a **Pythonic file interface**.
* Enables **typical file system operations** while storing connection metadata.
* Supports large-scale data operations using **PyArrow**.

---

## Semantic Data Types in ADS

ADS automatically detects **semantic types** after loading a dataset:

| Type            | Description                      | Examples                                 |
| --------------- | -------------------------------- | ---------------------------------------- |
| **Categorical** | Labeled groups, non-quantitative | Shirt size, eye color                    |
| **Ordinal**     | Ordered categories               | Education level (High School → Graduate) |
| **Continuous**  | Quantitative, measurable values  | Height, software version                 |
| **Datetime**    | Date and time formats            | 2025-10-05 08:30:00                      |

Inspect detected types:

```python
df.feature_types
df.show_in_notebook()
```

---

## Supported Sources and Formats

ADS supports a wide variety of **sources and formats** on OCI Data Science, including:

* **CSV, Parquet, JSON, ORC, Avro**
* **Databases** (Autonomous DB, MySQL, PostgreSQL, etc.)
* **Object storage systems** (OCI, AWS S3)
* **Big data interfaces** (PyArrow, HDFS)

Not supported:

* **DOC**, **PDF**, **Raw images**, **Text**, **Lists**, **Tuples**, **Ranges**.

> 🧩 A text extraction module is available to convert PDF/DOC into plain text.

---

## Summary

* Data can come from many structured and unstructured sources.
* ADS provides **efficient, secure, and optimized methods** to access data.
* Always choose the right connection method depending on:

  * Data source
  * Format
  * Access control (API key vs resource principal)
* Understanding the **semantic data types** is key for correct data preprocessing.
* Proper indexing, optimized SQL, and minimal latency improve performance.

---
