## 🧱 Model Deployment Overview

After completing **model training and evaluation**, the best candidate models are captured in the **Model Catalog**.  
Depending on your use case, models can be deployed for:
- **Batch consumption** – Predictions are scheduled periodically (e.g., hourly or daily).  
- **Real-time consumption** – Predictions are triggered immediately upon request.

### Example
Detecting fraudulent transactions in real-time payments requires **real-time prediction** through HTTP-based model deployment.

Model deployments in OCI are **managed resources** that allow you to deploy ML models as **HTTP endpoints** to serve real-time predictions.

---

## 🧩 OCI Model Deployment Architecture

### 🔗 Architecture Overview
The **client application** sends requests to a deployed model endpoint via API calls.  
Deployed models originate from the **Model Catalog** and run within a **Python runtime environment**, along with all required dependencies packaged in a **Conda environment**.

The architecture can be optionally integrated with **OCI Logging Service** for real-time inspection and debugging.

---

## ⚙️ Key Components of Model Deployment

| Component | Description |
|------------|-------------|
| **Load Balancer** | Distributes incoming traffic to multiple model servers. |
| **VM Instances Pool** | Hosts the model server, Conda environment, and the model itself. |
| **Model Artifact** | Contains the model file and prediction code (e.g., `score.py`). |
| **Conda Environment** | Includes all dependencies such as NumPy, Dask, or XGBoost. |
| **Logs** | Capture and emit inference logs to OCI Logging for monitoring and debugging. |

## 🚀 Creating and Invoking Model Deployments

A model entry in the **Model Catalog** includes two main files:
1. **`score.py`** – Defines inference logic (how the model makes predictions).  
2. **`runtime.yaml`** – Specifies the model’s runtime environment and deployment parameters.

Other configurations include:
- Compute shape,  
- Number of VM instances,  
- Logging setup,  
- Load balancer settings.

---

## 🧮 Example: Load Balancer Bandwidth Estimation

When estimating load balancer bandwidth, Oracle recommends **adding 20% overhead** for traffic spikes.

**Formula:**

\[
\text{Bandwidth (Mbps)} = \text{Payload (KB)} \times \text{Requests per second} \times 8 / 1024 \times 1.2
\]

**Example:**
If payload = 1024 KB and requests = 120/sec,  
then bandwidth = 1024 × 120 × 8 / 1024 × 1.2 = **1152 Mbps**.

## 🧰 Deployment Methods

You can deploy models using any of the following interfaces:

### 1. **OCI Console**
- Enter deployment name.  
- Select the model from the catalog.  
- Choose compute shape, instance count, logging, and bandwidth.

### 2. **ADS SDK**
- Use the `.deploy()` method of the **ModelDeployment** class.  
- Either create a **ModelDeploymentProperties** object and pass it in,  
  or define deployment parameters directly in the `.deploy()` call.

### 3. **OCI CLI**
- Run the `oci data-science model-deployment create` command.  
- Provide configuration details in a **JSON file**.  
- Optionally include a **log configuration file** for inference logging.

Once the deployment status is **active**, the endpoint can serve predictions.

## 🔗 Invoking Model Deployments

To invoke a model:
- Send **HTTP requests** to the model’s **predict endpoint**.  
- The model returns predictions as **HTTP responses**.

You can use:
- **OCI CLI**,  
- **OCI Python SDK**, or  
- **Java SDK** to invoke the deployed model.

> Note: OCI console provides sample code snippets for each invocation method.

### ⚠️ Important Constraints
- **Payload size limit:** 10 MB → Exceeding this returns a **413 HTTP error**.  
- **Timeout:** 60 seconds (fixed, cannot be changed).  
- For lower latency, use **streaming inference** with **Base64 encoding**.

## 🧭 Managing Model Deployments

You can **view, edit, and manage** model deployments from the OCI **Project Console**.

The console provides details such as:
- OCID,  
- Compute shape,  
- Logs,  
- Metrics, and  
- Work request status.

### Actions You Can Perform
- **Invoke** the predict endpoint.  
- **Edit** configuration parameters (e.g., name, compute shape, instance count).  
- **Apply changes** without downtime (if active).  
- **Modify all settings** when inactive.

---

## 🔄 Deactivating and Reactivating Deployments

- **Deactivating** a model stops VM instances and load balancer metering.  
  - Endpoints become unavailable.  
  - Metadata remains saved.  
- **Reactivating** restores the same HTTP endpoint and resumes serving predictions.  
- **Deleting** permanently removes the deployment from the project.

### OCI Logging

- **Prediction Logs** capture output and inference process messages from `score.py`.

- **Request count**

- **Bandwidth**

You can use:
