
# Lesson: Using AI Quick Actions in OCI Data Science  
### Rapid Deployment of Large Language Models (LLMs)

---

## 🧭 Introduction

**AI Quick Actions** is a recently introduced feature in **Oracle Cloud Infrastructure (OCI) Data Science**.  
It allows you to **quickly deploy pre-trained large language models (LLMs)** into your OCI Data Science environment.  

These deployed models can be:
- Integrated directly with applications.
- Fine-tuned using your **custom datasets**.
- Evaluated against **benchmark datasets** to measure performance.

This feature enables data scientists and developers to leverage **enterprise-grade LLMs** with minimal configuration effort.

---

## ⚙️ Key Capabilities

AI Quick Actions in OCI Data Science provide three main capabilities:

1. **Model Deployment**  
   Quickly deploy out-of-the-box LLMs (e.g., Mistral, Llama, etc.) within OCI Data Science.  

2. **Fine-Tuning**  
   Use your own dataset to fine-tune existing models for domain-specific applications.  

3. **Evaluation**  
   Evaluate the performance of deployed or fine-tuned models on specific datasets.  

---

## 🧩 Step 1: Accessing AI Quick Actions in OCI Data Science

AI Quick Actions can be accessed directly from an **OCI Data Science notebook session**.

Before you begin, ensure that:
- Your tenancy has the **required IAM policies** configured.  
- You have access to an **OCI Data Science notebook** session.  

Refer to the **OCI Data Science Documentation** for instructions on **enabling AI Quick Actions** through proper policies.

---

## 🧭 Step 2: Navigating the Interface

Inside your **OCI Data Science notebook**, locate and click on the **AI Quick Actions** button.  
This will open the **AI Quick Actions** dashboard, which contains **three tabs**:

1. **Models Tab** – Displays all available **out-of-the-box LLMs** ready for deployment.  
2. **Deployments Tab** – Displays all **active and past deployments**.  
3. **Evaluations Tab** – Used to **evaluate** deployed or fine-tuned models.

---

## 🧠 Step 3: Exploring Available Models

In the **Models tab**, you can browse through a list of pre-trained **Large Language Models** (LLMs) available for deployment.  
Each model includes information such as:
- Model name and version  
- Architecture type (e.g., Mistral 7B, Llama 2, etc.)  
- Training details and token limits  

These models are **optimized for OCI** and ready for one-click deployment.

---

## 🚀 Step 4: Creating a Deployment

Let’s go through the process of **deploying a model** using AI Quick Actions.

### Steps:
1. Go to the **Deployments** tab.  
2. Click on **Create Deployment**.  
3. Provide a **name** for your deployment.  
4. Select the **model** to deploy — for example,  
   **Mistral 7B Instruct (version 0.02)**.  
5. Choose a **compute shape**, e.g.:
```

VM.GPU.A10.1 or VM.GPU.8NR1.1

```
6. Select a **log group** for monitoring and troubleshooting.  
7. Click the **Deploy** button.

The deployment process begins automatically and takes a few minutes.

---

## 🔄 Step 5: Viewing and Managing Deployments

Once deployed, the model appears in the **Deployments** tab.  
Each deployment entry includes:

- **Deployment Name**  
- **Status** (e.g., Creating, Active, Failed)  
- **Endpoint URL** (for API access)  
- **Compute Shape Details**  
- **Logging Information**

### Example:
After deployment, you might see a model named:
```

mistral-7b-instruct-demo
Status: Active

```

When the model status changes to **Active**, it is ready for use.

---

## 🌐 Step 6: Using the Model Endpoint

Each deployed model provides an **API endpoint**.  
This endpoint can be used to **integrate the model** with your applications or external systems.

Example usage scenarios:
- Integrate with chatbots or web applications.
- Perform text summarization or question answering.
- Generate content dynamically using the LLM API.

The endpoint can be called via REST API or through OCI SDKs (Python, Java, etc.).

---

## 🧪 Step 7: Testing the Model in the Notebook

You can **test** the deployed LLM directly within the **AI Quick Actions** page.

### Steps:
1. Open your active deployment.  
2. Locate the **Test Section**.  
3. Adjust model parameters (e.g., temperature, max tokens, top_p).  
4. Enter a **prompt** in the text box.  
   Example prompt:
```

Tell us about Las Vegas.

```
5. Click on the **Generate** button.

The model processes the prompt and returns a response in real-time, which appears on the screen.

---

## ⚙️ Step 8: Fine-Tuning and Evaluation

Beyond deployment, AI Quick Actions also support:

- **Fine-tuning LLMs** using your custom dataset.  
- **Evaluating** models using test datasets to assess accuracy and relevance.

These steps are available through the **Evaluations tab**, where you can:
- Upload datasets.  
- Configure evaluation parameters.  
- View and compare results across models.
