## 🧠 What Are ADSEvaluators?

ADSEvaluators help in **calculating and visualizing** model performance metrics automatically.  

There are **three main types** of ADSEvaluators in ADS:

1. **Binary Classifier Evaluator**  
   For tasks where the output is binary (e.g., 0 or 1, Yes or No).

2. **Multinomial (Multiclass) Classifier Evaluator**  
   For tasks involving multiple categories or classes.

3. **Regression Evaluator**  
   For predicting continuous numerical values (e.g., prices, temperatures).

Each evaluator automatically computes standard metrics and generates visual plots to assess model quality.

---

## 🧩 Why Use ADSEvaluators?

ADSEvaluators drastically simplify the evaluation process by:

- Automatically computing all key metrics.  
- Generating comparison results for multiple models simultaneously.  
- Providing visual outputs like **confusion matrices**, **ROC curves**, and **precision-recall plots**.  
- Allowing quick iteration for model improvement and tuning.

## 🧮 Example: Binary Classification with ADSEvaluator

Let’s walk through a simple example to demonstrate how to use the **ADSEvaluator** class in a binary classification scenario.

### Step 1: Import Required Libraries
```python
from ads.evaluations import ADSEvaluator, ADSModel
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
````

---

### Step 2: Create and Split the Dataset

```python
# Generate a binary classification dataset
X, y = make_classification(n_samples=1000, n_features=10, random_state=42)

# Split into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

---

### Step 3: Train Multiple Models

```python
# Create and fit estimators
lr = LogisticRegression()
rf = RandomForestClassifier()

lr.fit(X_train, y_train)
rf.fit(X_train, y_train)
```

---

### Step 4: Wrap Estimators with ADSModel

```python
# Wrap models with ADSModel class
lr_model = ADSModel.from_estimator(lr)
rf_model = ADSModel.from_estimator(rf)
```

---

### Step 5: Create the Evaluator Object

```python
# Create evaluator for both models
evaluator = ADSEvaluator(
    models=[lr_model, rf_model],
    X=X_test,
    y=y_test
)
```

---

### Step 6: Print Evaluation Metrics

```python
# Display performance metrics
evaluator.metrics
```

This prints all relevant metrics for each model, such as **accuracy**, **precision**, **recall**, **F1 score**, and **ROC AUC**, for both **training** and **testing** datasets.

---

### Step 7: Plot Evaluation Graphs

```python
# Display visual charts and metrics in the notebook
evaluator.show_in_notebook()
```

This method generates key plots such as:

* **Confusion Matrix**
* **ROC Curve**
* **Precision-Recall Curve**
* **Lift and Gain Charts**

---

## 📊 Example Output (Conceptual)

When running the example, you will see metrics printed like this:

| Model                  | Dataset | Accuracy | Precision | Recall | F1 Score | ROC AUC |
| :--------------------- | :------ | :------- | :-------- | :----- | :------- | :------ |
| LogisticRegression     | Test    | 0.89     | 0.91      | 0.87   | 0.89     | 0.90    |
| RandomForestClassifier | Test    | 0.92     | 0.93      | 0.91   | 0.92     | 0.95    |

Below these metrics, **visual graphs** will appear, illustrating the model performance for deeper insights.

---

## 💡 Key Takeaways

* **ADSEvaluators** automate the process of **calculating and plotting model metrics**.
* You can easily compare **multiple models** using a single evaluator instance.
* The **evaluator.metrics** command displays key metrics.
* The **evaluator.show_in_notebook()** command plots relevant charts (e.g., confusion matrix).
* They make evaluating and visualizing models **fast, consistent, and intuitive**.

---

## 🧭 Hands-on Practice

To reinforce your learning:

1. Try creating your own dataset using `make_classification()`.
2. Train different classifiers (e.g., SVM, DecisionTree).
3. Use `ADSEvaluator` to compare results.
4. Observe how metrics and plots differ for each model.

---
