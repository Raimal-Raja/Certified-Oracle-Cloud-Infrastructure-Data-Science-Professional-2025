## 🧠 About Hugging Face Pipelines

The **Hugging Face Transformers library** offers pre-trained models that can be easily integrated into production pipelines.  
Recently, OCI Data Science added native **support for deploying Hugging Face pipelines**, allowing seamless deployment of transformer-based models.

With Hugging Face pipelines, you can perform tasks such as:

- Text classification  
- Sentiment analysis  
- Named entity recognition  
- Question answering  
- Translation and summarization  
- Image classification and object detection  

---

## ⚙️ Step 1: Importing Required Libraries

We start by importing the necessary modules from the **Transformers** library.

```python
from transformers import pipeline
from ads.model import HuggingFacePipelineModel
````

The `pipeline` function helps build a ready-to-use NLP or vision model, while the `HuggingFacePipelineModel` class wraps the pipeline for OCI Data Science deployment.

---

## 🧩 Step 2: Building the Hugging Face Pipeline

Create a simple transformer-based pipeline.
For example, let’s build a **sentiment analysis** pipeline using a pre-trained model:

```python
nlp_pipeline = pipeline("sentiment-analysis")
```

This automatically downloads the model and tokenizer required for performing sentiment analysis.

---

## 🪄 Step 3: Wrapping the Pipeline

Now we wrap our Hugging Face pipeline with the **`HuggingFacePipelineModel`** class:

```python
hf_model = HuggingFacePipelineModel(pipeline=nlp_pipeline)
```

This step prepares the Hugging Face model to interact with OCI Data Science services.

---

## 🧰 Step 4: Preparing Model Artifacts

We prepare the model artifacts using the `.prepare()` method.
This method packages all necessary files (model weights, configurations, and scripts) for deployment.

```python
hf_model.prepare(inference_conda_env="huggingface_env")
```

---

## ✅ Step 5: Verifying the Model

Before saving or deploying, it’s important to verify the model using the `.verify()` method:

```python
hf_model.verify()
```

This ensures that the model and its artifacts are correctly structured and ready for deployment.

---

## 💾 Step 6: Saving the Model to the Catalog

Once verification is complete, save the model into the **Model Catalog** using the `.save()` method:

```python
hf_model.save()
```

This registers the model in your OCI Data Science Model Catalog for tracking and versioning.

---

## 🚀 Step 7: Deploying the Model

Finally, deploy your Hugging Face model as a REST API endpoint using the `.deploy()` method:

```python
deployment = hf_model.deploy(display_name="HuggingFaceSentimentModel")
```

The deployment process automatically provisions compute resources and exposes the model as a secure endpoint that can be invoked via HTTP, SDK, or CLI.

---

## 📘 Summary

Let’s recap the steps for deploying a Hugging Face pipeline in OCI Data Science:

1. **Import** the Transformers and ADS libraries
2. **Build** a Hugging Face pipeline
3. **Wrap** it using `HuggingFacePipelineModel`
4. **Prepare** model artifacts with `.prepare()`
5. **Verify** deployment readiness using `.verify()`
6. **Save** the model in the Model Catalog using `.save()`
7. **Deploy** the model with `.deploy()`

---
