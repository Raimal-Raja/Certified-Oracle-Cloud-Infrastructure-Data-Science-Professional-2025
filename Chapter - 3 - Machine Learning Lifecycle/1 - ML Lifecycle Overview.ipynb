## 1. Introduction

Welcome to the module covering the **Machine Learning (ML) Lifecycle**.  
In this lesson, we’ll explore an overview of the ML lifecycle and its six major steps.  

Organizations need **versatile and productive tools** to support every stage of the ML process. The smoother and more automated this lifecycle is, the more valuable it becomes — because faster and more frequent results mean greater impact.

The six key stages of the ML lifecycle are:

1. Data Access  
2. Data Exploration and Preparation  
3. Modeling (Building and Training)  
4. Model Validation  
5. Model Deployment  
6. Model Monitoring  

> The lifecycle always starts with a **business problem** that provides direction and objectives.

---

## 2. Overview of the Lifecycle

Machine learning is an **iterative process** — data scientists repeat many steps to refine performance and improve accuracy.

While there are more complex frameworks such as **CRISP-DM (Cross Industry Standard Process for Data Mining)**, this simplified version is sufficient for understanding and implementing ML projects within this module.

---

## 3. Step 1: Data Access

Every machine learning project begins with **data**.

- In **OCI Data Science**, data is typically stored in the **notebook session** for quick access.  
- Data can come from:
  - Enterprise data management systems  
  - Data lakes and relational or non-relational databases  
  - Unstructured sources like logs, text, images, or video  
  - Open-source or public datasets  
  - Purchased data or collected data (surveys, clickstreams, sensors, cameras)  

> Understanding the organization’s **data ecosystem** helps identify valuable data sources.  
> The **OCI Data Catalog** can assist in locating existing datasets.

---

## 4. Step 2: Data Exploration and Preparation

After collecting data, data scientists must **explore, clean, and transform** it.

### Key Activities:
- Visualize and understand the data structure.  
- Identify and handle:
  - Duplicates  
  - Inaccuracies  
  - Missing or corrupt values  
  - Outliers  
- Determine whether data is **labeled** (supervised) or **unlabeled** (unsupervised).  
  - Example: In image detection, labeled data includes bounding boxes identifying objects.

Oracle provides **OCI Data Labeling Cloud Service** to assist with labeling tasks.

### Exploratory Data Analysis (EDA)
- Examine feature types and relationships.
- Calculate summary statistics (min, max, mean, median, mode).
- Analyze data distributions and correlations.
- Visualize data through plots to identify:
  - Skewed distributions  
  - Biases  
  - Missing or extreme values  

### Feature Engineering
Create new features that better represent the data.  
Example: Transform “hour” data into categories like *early morning, afternoon, evening*.

Categorical features often require **one-hot encoding** or similar transformations to make them usable for modeling.

---

## 5. Step 3: Modeling (Building and Training)

Modeling involves selecting and training the appropriate algorithms to solve a problem.

### Key Steps:
1. **Choose model type:**
   - **Supervised Learning**: Has labeled data (e.g., regression, classification).
   - **Unsupervised Learning**: No labels (e.g., clustering, segmentation).  

2. **Experimentation:**
   - Try multiple algorithms.
   - Test different feature subsets.
   - Split data into **training** and **testing** sets.

3. **Training:**
   - Train the model on the training set.
   - Evaluate on the test set to assess generalization.

### Benefits of Reducing Input Variables:
- Lowers computational cost.
- Improves generalization.
- Simplifies model interpretation.

---

## 6. Step 4: Model Validation and Evaluation

After training, the model must be evaluated to ensure it meets performance and business objectives.

### Common Metrics:

**For Classification:**
- Accuracy  
- Precision  
- Recall  
- F1 Score  
- Confusion Matrix (True/False Positives/Negatives)

**For Regression:**
- Root Mean Square Error (RMSE)  
- Mean Absolute Error (MAE)  
- R² (Coefficient of Determination)

**For Unsupervised Models:**
- High cohesion and separation between clusters.

> The choice of metrics depends on the business context.  
> Example: In rare disease detection, **recall** is more important than overall accuracy.

---

## 7. Step 5: Model Deployment

Model deployment makes the trained model available for use.

### Deployment Modes:
- **Batch Inference**: Predictions at scheduled intervals (hourly, daily).
- **Real-Time Inference**: Instant predictions triggered by live data (e.g., fraud detection).

### Considerations:
- Response time requirements (milliseconds vs. seconds)
- Request volume and data size
- Resource usage (memory, CPU)
- Integration with pipelines and applications

> Data scientists often collaborate with **MLOps engineers** to operationalize models.

---

## 8. Step 6: Model Monitoring

After deployment, models must be monitored continuously to ensure reliability and accuracy.

### Components of Monitoring:

1. **Drift Monitoring**
   - Detects changes in data distributions or prediction patterns.  
   - Compares live data with training data to identify shifts.  
   - Detects model degradation over time.

2. **Operational Monitoring**
   - Tracks latency, CPU/memory usage, throughput, and system reliability.  
   - Uses logs and metrics to identify incidents and root causes.

### Why It Matters:
- Ensures sustained model performance.  
- Detects concept drift early.  
- Determines when to retrain and redeploy models.

Machine learning is **highly iterative** — data scientists often loop through these steps multiple times to refine results and meet objectives.
