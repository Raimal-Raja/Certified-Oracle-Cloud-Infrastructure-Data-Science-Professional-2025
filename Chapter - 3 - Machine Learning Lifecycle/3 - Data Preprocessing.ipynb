## 2. Why Preprocessing is Needed

Real-world data is often:
- Incomplete, inconsistent, or noisy.
- Contains missing values, errors, or outliers.

Therefore, preprocessing ensures:
- Data consistency and quality.
- Better model training and generalization.

> Preprocessing is typically the **largest part** of the entire ML lifecycle.

---

## 3. Data Operations and Integration

Data often comes from multiple sources and needs to be **combined**.

Common operations include:
- **Appending**, **deleting**, and **filtering** rows/columns.  
- **Joining** on columns or indexes.  
- **Concatenating** datasets vertically or horizontally.

All operations available for **pandas DataFrames** can also be applied to **ADS datasets**.

Ensure:
- Consistent **formats**, **units**, and **naming conventions**.
- Checking **statistical summaries**, **data types**, **null values**, and **duplicates**.

---

## 4. Data Imputation

**Imputation** deals with missing data caused by:
- Human errors (e.g., incorrect data entry).
- Transmission errors (e.g., GPS failures).

### Common methods:
- **Deletion** of incomplete rows (least recommended).
- **Filling** missing data with:
  - Mean or median (for numerical data).
  - Mode (for categorical data).

### Encoding Categorical Data:
- **Label Encoding:** Converts categories into numbers (useful for nominal data).
- **One-Hot Encoding:** Transforms categorical columns into dummy variables (better for ordinal data).

ADS provides built-in modules like `label_encoder` and methods such as `fit_transform`.

---

## 5. Outlier Detection

**Outliers** are data points that deviate significantly from other observations.

They may represent:
- Errors, or
- Genuine rare events.

### Detection methods:
- **Visualization:** Scatter plots, box plots.
- **Statistical measures:** Based on Gaussian (normal) distributions.
- **Machine Learning:**  
  - **Supervised methods** (classification-based, need labeled data).  
  - **Unsupervised methods** (no labels, assumes outliers are rare).

Unsupervised methods are often preferred since they handle unlabeled data effectively.

---

## 6. Feature Scaling

Feature scaling adjusts the **scale of features** so that all contribute equally to the model.

### Two main approaches:
1. **Normalization (Min-Max Scaling):**  
   Scales all features to the range **[0, 1]**.
2. **Standardization (Variance Scaling):**  
   Subtracts the mean and divides by the variance — results in mean = 0 and standard deviation = 1.

These are essential when using algorithms sensitive to distance (e.g., regression).

---

## 7. Dimensionality Reduction

**Dimensionality** = number of features or input variables.

High-dimensional data:
- Increases computational cost.
- Makes model control difficult.

### Two main techniques:
1. **Feature Selection:** Choose the most important subset of features.
2. **Feature Extraction:** Derive new features from existing ones.

> Full exploration of these methods is beyond the scope of this course.

## 8. Text Data Preprocessing

Text preprocessing differs from numerical preprocessing. It may include:
- Vectorization
- Stop-word removal
- POS tagging
- Tokenization
- Stemming
- Lemmatization

ADS provides built-in tools to simplify these transformations.

## 9. ADS Transformation Tools

ADS provides powerful utilities for data transformation and optimization:

### a. `suggest_recommendations`
- Detects issues with data.
- Recommends transformations (e.g., fix missing values, handle imbalance).
- User can accept recommendations through a simple interface.

### b. `auto_transform`
- Applies all recommended transformations automatically.
- Handles:
  - Missing value imputation.
  - Noise removal.
  - Correlation reduction.
  - Class imbalance (via upsampling/downsampling).
  - Dropping constraint and primary key columns.

> Produces a new optimized dataset ready for modeling.

### c. `visualize_transforms`
- Displays a flow diagram of automated transformations.
- Helps users understand how data has been modified.
- Does not include custom transformations.

## 11. Splitting Data Sets

**Splitting data** into **train**, **test**, and **validation** sets is a vital step in ML.

### Default ADS Split:
- **Training:** 80%  
- **Testing:** 10%  
- **Validation:** 10%

### Notes:
- Larger test/validation sets may waste valuable data.
- For limited data:
  - Use **70%** or even **60%** for training to retain enough test examples.

This ensures a fair estimate of model generalization on unseen data.
