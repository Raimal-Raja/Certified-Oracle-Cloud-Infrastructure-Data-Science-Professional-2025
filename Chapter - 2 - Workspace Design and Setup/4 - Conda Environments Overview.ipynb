# Conda Environments in Oracle Cloud Infrastructure Data Science Service

**Speaker:** John Peach, Data Scientist at Oracle Cloud Infrastructure  

---

## 1. Introduction
- Topic: **Conda environments in the Data Science Service**

---

## 2. What is a Conda Environment?
- Open-source **package & environment manager**  
- Bundles:
  - Python interpreter  
  - Modules  
  - Libraries  
  - Other programs  

---

## 3. Benefits of Conda Environments
- Install only required packages.  
- Isolate different configurations (e.g., CV vs. regression).  
- Switch between environments easily.  
- Share configurations with team members.  
- Deploy models with the same environment they were trained in.  
- Supports **reproducible research** (specific package versions).  

---

## 4. Environment Explorer
- **GUI tool** for managing conda environments.  
- Views: 
  - **Card view** (details)  
  - **List view** (overview)  
- Features:  
  - Search  
  - Filtering (e.g., GPU-enabled, deprecated)  

---

## 5. Types of Conda Environments
### 1. Data Science Conda Environments
- Managed by Oracle DS Service team.  
- Curated for tasks:
  - PyTorch  
  - TensorFlow  
  - Healthcare  
  - General-purpose (EDA).  

### 2. Published Conda Environments
- Created & managed by customers.  
- Stored in **Object Storage**.  
- Share across teams, notebooks, jobs, deployments.  

### 3. Installed Conda Environments
- Specific to a **notebook session**.  
- Can be Data Science or Published environments.  
- Persist on **block volume** → available across sessions.  

---

## 6. Recap
- **Conda environments = containers for software collections**  
- Benefits: isolation, reproducibility, sharing, deployment consistency.  
- Environment Explorer helps discover, manage, and filter environments.  
- 3 Types:  
  1. Data Science (managed)  
  2. Published (customer-created)  
  3. Installed (per session)
