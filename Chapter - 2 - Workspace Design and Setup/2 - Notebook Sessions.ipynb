
## Introduction  
- **Notebook sessions** are the JupyterLab interface for building and training models.  
- Fully managed by the service (no need to handle compute or storage APIs).  
- Software updates and patching are automatic.  
- Support for **CPU and GPU shapes**.  
- Provide **persistent storage** for data, notebooks, and environments.  



## Creating a Notebook Session  
Steps to create from the console:  
1. Navigate to your **project details page**.  
2. Click **Create notebook session**.  
3. Select your **compartment**.  
4. Optionally, give the session a **name**.  
5. Choose a **compute shape**:  
   - AMD CPUs  
   - Intel CPUs  
   - Nvidia GPUs  
6. Specify **block storage size** (50 GB – 10,240 GB).  
7. Select **networking type**:  
   - Default networking  
   - Custom networking (choose VCN + subnet).  
8. Click **Create**.  
   - Takes a few minutes to start.  
   - When active → **Open** button is enabled.  



## Managing Notebook Sessions  

- **View Sessions**:  
  - From the notebook sessions list page.  
  - Metadata includes:  
    - Display name + OCID  
    - Date/time created + user  
    - Compute shape + storage size  
    - VCN + subnet  
    - Tags  

- **Edit Session**:  
  - Only field editable when active → **Display Name**.  
  - Changes via any OCI interface.  

- **Delete Session**:  
  1. Locate session in project details page.  
  2. Click **Delete**.  
  3. Enter session display name to confirm.  
  4. Compute + block volume destroyed.  
  - ⚠️ **Uncommitted files are lost**.  



## Activating & Deactivating Sessions  

### Activate:  
- From details/list page → click **Activate**.  
- Select/confirm compute, storage, network settings.  
- Can **scale up or down** depending on requirements.  
- Status:  
  - Updating → Active (few minutes).  

### Deactivate:  
- From details/list page → click **Deactivate**.  
- Compute instance shut down.  
- Block volume retained (data/code safe if saved there).  
- ⚠️ Data on **boot volume** is deleted.  
- On reactivation → new compute instance + same block volume.  



## Metrics  

You can view metrics for a session on its **Details page**:  
- CPU utilization (% of total).  
- Memory utilization (% of total).  
- Network traffic (bytes in/out).
