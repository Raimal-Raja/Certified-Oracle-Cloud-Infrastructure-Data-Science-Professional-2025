1. Introduction

Speaker: John Peach, Data Scientist at Oracle Cloud Infrastructure Data Science Service.

Focus: Data Science Service Conda Packs – pre-made environments curated by Oracle.

2. What is a Data Science Service Conda Pack?

Pre-built conda environments → easier than creating from scratch.

Based on open-source software, customizable for specific needs.

Accessible via Jupyter Lab Launcher → Environment Explorer.

All Oracle DS Service environments include:

OCI Python SDK

ADS SDK (AutoML & Explainability in select environments).

3. Types of Conda Environments

Application-based (e.g., ONNX, PyPGX, PySpark).

Use case-based (e.g., computer vision, data exploration, finance).

Each comes with best-of-breed software tailored to the task.

4. Conda Environment Families

Grouped by Python version + Architecture (CPU/GPU).

Example:

NLP for CPU on Python 3.7 v2 vs. NLP for CPU on Python 3.7 v1 → software changes, same base.

Naming conventions:

Application-based: <Software> <Version> for <Arch> on Python <Ver> (e.g., PyTorch 1.10 for GPU on Python 3.7).

Use-case based: <Task> for <Arch> on Python <Ver> (e.g., Data Exploration for CPU on Python 3.7).
