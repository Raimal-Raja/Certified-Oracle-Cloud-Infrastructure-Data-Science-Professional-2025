## Module Overview
In this module, we will explore version control systems (VCS), their importance in data science workflows, and how Git integrates with Oracle Cloud Infrastructure (OCI) Data Science Service.

---

## 1. Introduction to Version Control Systems
- Also known as **Source Code Management (SCM)** systems.  
- Tools for managing **different versions** of code, documents, data, and analyses.  
- Originally developed for software engineering — now widely adopted in data science.

### Examples of SCM Tools:
- CVS, Bazaar, Subversion, Perforce, CodeCommit, Mercurial, Helix Core, etc.  
- **Git** is the most popular and is integrated into OCI Data Science Service.

### Usage in Data Science:
- Tracking Jupyter Notebooks, code, reports, and data.  
- Managing analysis workflows.  
- Enabling collaboration and change management.  

---

## 2. Key Concepts

### Repository (Repo)
- A **directory** that contains all changes for a project or analysis.  
- Functions like a **filing cabinet** holding versions of software or analysis.  
- Each repo stores a **codebase** separate from others.  

### What Repos Do:
- Act as a **central place** for storing and sharing code.  
- Provide **version control** — track all development versions.  
- Enable **merging** of multiple contributors' changes.  
- Support **branching** to isolate changes for testing or feature development.  

---

## 3. Benefits of Version Control for Data Scientists
- Maintain **reproducibility** of experiments.  
- Allow collaboration on shared analyses.  
- Archive historical versions.  
- Easily **revert** to previous working code.  
- Enable **parallel work** via branching and merging.  

**Example:**  
If you develop a working ML algorithm and later experiment with improvements that fail, you can revert to the last working version because you committed it earlier.  

---

## 4. Centralized vs Distributed Systems

### Centralized VCS
- Relies on a **central server** for all commits.  
- Easier setup and centralized admin control.  
- Examples: Subversion, CVS, Perforce.  

### Distributed VCS
- **No single point of failure.**  
- Developers clone repositories locally.  
- Work offline and push/pull when ready.  
- Examples: **Git**, Bazaar, Mercurial.  

### Hybrid Approach
- Many teams use Git in a **centralized configuration** (e.g., GitHub, OCI Code Repo) for collaboration.  

---

## 5. Git in Data Science Workflow
- Tracks changes to code and notebooks.  
- Enables reverting, merging, and versioning.  
- Facilitates collaboration via shared repositories.  
- Fast and fault-tolerant — local commits work offline.  

### Git in OCI Data Science
- Git extension in JupyterLab for point-and-click Git operations.  
- Works with GitHub, Bitbucket, GitLab, and OCI Code Repositories.  
- Simplifies version tracking for notebook sessions.  

---

## 6. Key Git Terminology
**Commit:** A snapshot of the project at a point in time, identified by a SHA (unique 40-character hash).  

**Repository (Repo):** A directory containing all project commits, branches, and files.  

**Working Area:** The current workspace with the project’s files.  

**Staging:** Selecting modified files to include in the next commit.  

---

## 7. Git Workflow Summary
1. Modify files in the **working area**.  
2. **Stage** changes (`git add`).  
3. **Commit** changes (`git commit -m "message"`).  
4. **Push** commits to a remote repo.  
5. **Pull** changes to update your local repo.  

Commits are lightweight, so it’s best to commit early and often.  

---

## 8. OCI Code Repository
- Acts as a central **Git-compatible** peer.  
- Integrated with OCI **Identity and Access Management (IAM)**.  
- Allows commits, branches, and updates.  
- Each repo has a unique **OCID** identifier.  
- You can connect external repos (GitHub, GitLab, etc.) securely via the **OCI Vault** for token management.  

---

## 9. Connecting to GitHub via SSH

### Steps:
1. Install and configure Git (set name & email).  
2. Create a **GitHub account**.  
3. Generate an **SSH key pair** (public & private).  
4. Add the **public key** to GitHub.  
5. Create local and remote repos.  
6. Clone the remote repo (`git clone`).  
7. Add and commit files locally.  
8. Push changes to GitHub (`git push`).  

---

## 10. Common Git Commands

| Command | Description |
|----------|-------------|
| `git init` | Initialize a new repository |
| `git clone <URL>` | Copy a remote repo locally |
| `git add <file>` | Stage file(s) for commit |
| `git commit -m "message"` | Commit staged changes |
| `git push` | Upload commits to remote repo |
| `git pull` | Fetch & merge remote changes |
| `git fetch` | Download data from remote repo |
| `git remote` | Manage remote connections |

- **Two types:** Centralized and Distributed.
