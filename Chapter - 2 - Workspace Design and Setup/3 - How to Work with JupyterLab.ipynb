## Introduction
- JupyterLab is a next-generation web-based UI.  
- Used as the interface for notebook sessions.  
- Familiar for data scientists → easy adoption.  
- Supports integration of:
  - Jupyter Notebooks  
  - Text editors  
  - Terminals  
  - Custom components  
- Understands many file formats: images, CSV, JSON, Markdown, PDF, Vega, Vega Lite.  

---

## Differences in Data Science Console vs Open Source
Menu and layout are identical, but with 3 extra features:
1. **Launcher** – quick access to notebooks, console, terminal, text editor, Environment Explorer, examples.  
2. **Environment Explorer** – GUI tool to manage/search conda environments.  
3. **GitHub Extension** – enables Git version control inside sessions.  

---

## Key Features of JupyterLab
- **Menu Bar**: top-level menus with keyboard shortcuts.  
- **Launcher**: access to notebooks, console, terminals, editors, and extensions.  
- **Left Sidebar**:
  - File browser  
  - List of tabs  
  - Command palette  
  - Table of contents  
  - Extension manager  
- **Main Work Area**: arrange panels of tabs, resizable/subdividable.  
- **Code Consoles**: scratch pads for interactive code with rich output.  
- **Kernel Management**: accessible from kernel menu.  
- **Multiple Views**: live editing of documents across viewers.  
- **Right Sidebar**: Property Inspector (active in notebooks).  
- **Command Palette**: keyboard-driven command search.  

---

## Top Chrome Bar (Console UI)
- Oracle logo → back to Cloud Console home.  
- Session name → link to notebook session details.  
- Session remaining → auto-logout after 1 hour (extendable up to 24 hours).  
- Help icon → OCR docs.  
- Sign out → end notebook session.  

---

## Left Sidebar in Detail
- File browser (navigate, open, add, delete).  
- Terminal/kernel sessions → shut down processes.  
- Git extension (covered separately).  
- Commands tab → search commands (e.g., Variable Inspector).  
- Property Inspector.  
- Open tabs list.  
- Table of contents (markdown parsing).  
- Extension manager.  
- Hide sidebar → click icons twice.  

---

## Launcher
- Easy access to notebooks, terminals, editors, Environment Explorer, examples.  
- **Left Side**: Environment Explorer + notebook examples.  
- **Right Side**: create new files, sessions, or terminals.  
- Select kernel (e.g., Python3) → create notebook.  

---

## Working with Notebooks
- Rename notebooks via right-click → Rename.  
- Run code in cells:
  - Play icon  
  - Run menu → Selected Cell  
  - Shortcut: `Shift+Enter`  
- Cell types: Code, Markdown, Raw.  
- Reorder cells by dragging.  
- Edit menu options: Merge cells, Split cells.  
- Change kernel from top-right menu.  

---

## Example Notebook
- Load pre-installed notebook (e.g., binary classification attrition).  
- Run all cells, stop via Kernel → Interrupt.  
- Use Variable Inspector alongside notebook.  
- Drag tabs to split/merge views.

## View Menu Options
- Show/hide line numbers.  
- Collapse/expand code cells or outputs.

- Change theme.

- Example: upload CSV → displayed in table format.  
- Open terminal → run Linux + CLI tools (conda, git, OCI).
