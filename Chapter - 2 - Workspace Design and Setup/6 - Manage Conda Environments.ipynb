1. Introduction

Speaker: John Peach, Data Scientist, Oracle Cloud Infrastructure.

Topic: Managing Conda Environments with odsc CLI tool.

Key Idea: GUI (Environment Explorer) is simple, but CLI (odsc) provides more control.

2. Recap: What are Conda Environments?

Bundled collections of software (interpreter, libraries, packages).

In OCI:

Data Science Conda Packs (managed by Oracle team).

Published Conda Packs (created by user).

Managed via:

Environment Explorer (GUI).

odsc command line tool (full control).

3. Functionalities of odsc CLI
🔹 Browse

View environments in YAML format (name, description, libraries, slug).

Commands:

odsc conda list → all Data Science conda environments.

odsc conda list --local → only installed environments.

odsc conda list --override → published environments.

🔹 Search

No direct search command.

Combine odsc conda list with Unix tools (grep, awk, perl).

Example:

odsc conda list | grep -E "name|slug"

🔹 Install

Install Oracle-managed or published conda packs.

Command:

odsc conda install --slug <slug_name>


To install published envs: add --override.

🔹 Clone

Safely copy an environment before making changes.

Command:

odsc conda clone --from-environment <source_slug> --env <new_name>


Creates a new slug automatically.

🔹 Modify

Done using conda tool, not odsc.

Steps:

Activate environment:

conda activate /home/datascience/conda/<slug>


Upgrade package (e.g., ADS):

python3 -m pip install oracle-ads --upgrade

🔹 Publish

Share environments via Object Storage Bucket.

Steps:

Create bucket in OCI.

Initialize:

odsc conda init --bucket-namespace <namespace> --bucket-name <bucket>


Publish:

odsc conda publish --slug <slug_name>

🔹 Delete

Free up space by removing unused envs.

Command:

odsc conda delete --slug <slug_name>

🔹 Create from Scratch

Use manifest YAML file.

Command:

odsc conda create --file <manifest.yaml>


Defaults include /opt/base-env.yaml.

Use --empty to skip base packages.

4. Recap

odsc CLI = full-featured conda environment manager.

Key functions: browse, install, clone, modify, publish, delete, create.

Works with both Oracle-managed and user-published environments.

CLI power: integrate with Unix tools for advanced searches.
