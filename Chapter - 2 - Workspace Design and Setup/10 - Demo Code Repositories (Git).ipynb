## 1. Objective
To demonstrate how to:
- Create a **local Git repository**.
- **Commit files** to the repository.
- **Link a local repo** to a GitHub repository.
- **Push and pull** files between the local and remote repositories.

---

## 2. Git Configuration

1. Open a **Terminal** in JupyterLab.  
   (Git is pre-installed in the Data Science environment.)

2. Configure Git with your **user name** and **email**:
   ```bash
   git config user.name "Your Name"
   git config user.email "your.email@domain.com"
````

These values identify who made each commit.

3. Verify configuration:

   ```bash
   git config --list
   ```

   ✅ You should see your name and email listed.

---

## 3. Creating a Local Repository

1. Open the **File Browser** in JupyterLab.
2. Right-click → **New Folder** → name it `demo`.
3. Go to the **Git icon** in the left sidebar → click **Initialize Repository**.
4. Confirm initialization — you now have a **Git repository**.

You’ll now see Git-related tools available in the sidebar.

---

## 4. Setting Up SSH Authentication for GitHub

GitHub requires **SSH key authentication** to securely push changes.

### a. Generate SSH Key Pair

Run:

```bash
ssh-keygen -t rsa -b 4096 -C "your.email@domain.com"
```

* Choose default location when prompted.
* Press Enter for passphrase (optional).

This creates two files:

* **Private key** → keep secret.
* **Public key** → share freely.

Verify creation:

```bash
ls ~/.ssh
```

### b. View and Copy the Public Key

Display and copy the key:

```bash
cat ~/.ssh/id_rsa.pub
```

### c. Start SSH Agent

```bash
eval "$(ssh-agent -s)"
```

### d. Add Private Key to SSH Agent

```bash
ssh-add -k ~/.ssh/id_rsa
```

---

## 5. Adding SSH Key to GitHub

1. Go to your **GitHub Profile → Settings → SSH and GPG keys**.
2. Click **New SSH Key**.
3. Paste your **public key** into the text box.
4. Click **Add SSH Key**.
   ✅ You’ll now see the key added to your account.

---

## 6. Creating a GitHub Repository

1. On GitHub, click **New Repository**.
2. Enter name → `demo`.
3. Click **Create Repository**.
4. Copy the **SSH repository URL**, e.g.:

   ```
   git@github.com:username/demo.git
   ```

---

## 7. Linking Local Repository to GitHub

In your JupyterLab terminal:

```bash
git remote add origin git@github.com:username/demo.git
```

Verify the connection:

```bash
git remote -v
```

✅ You should see `origin` pointing to your GitHub repo (fetch/push).

---

## 8. Creating and Committing a Notebook

1. Go to **Launcher → File → New → Notebook**.
2. You’ll see it listed as **Untracked** in the Git sidebar.
3. Click to **Track** it.
4. Stage and commit:

   * Enter commit message: `Initial Commit`.
   * Click **Commit**.

---

## 9. Pushing the Commit to GitHub

1. Set the upstream branch (first time only):

   ```bash
   git push --set-upstream origin master
   ```
2. After that, you can simply push:

   ```bash
   git push
   ```

✅ Your notebook is now uploaded to GitHub.

---

## 10. Verifying the Upload

1. Go to your GitHub repository page.
2. You should see your committed notebook file and message:

   ```
   Initial Commit
   ```

---

## 11. Making and Committing Further Changes

1. Edit your notebook (e.g., add code: “some magical math”).
2. Save the notebook.
3. In the Git sidebar:

   * The file will appear as **Modified**.
   * Right-click → **Stage**.
   * Add commit message: `Added some math`.
   * Click **Commit**.
4. Push the update:

   ```bash
   git push
   ```
5. Check GitHub again → refreshed changes appear.

---

## 12. Workflow Recap

| Step | Action                             |
| ---- | ---------------------------------- |
| 1    | Configure Git (name, email)        |
| 2    | Initialize local repository        |
| 3    | Generate SSH key pair              |
| 4    | Add public key to GitHub           |
| 5    | Create GitHub repository           |
| 6    | Link local → remote repo           |
| 7    | Create, track, and commit notebook |
| 8    | Push commits to GitHub             |
| 9    | Make updates and push again        |

---

## 13. Key Commands Summary

| Command                       | Description                   |
| ----------------------------- | ----------------------------- |
| `git config --list`           | View Git settings             |
| `git init`                    | Initialize a new repository   |
| `ssh-keygen`                  | Generate SSH keys             |
| `ssh-add`                     | Add private key to SSH agent  |
| `git remote add origin <URL>` | Link local repo to remote     |
| `git status`                  | Show current changes          |
| `git add <file>`              | Stage file for commit         |
| `git commit -m "message"`     | Commit changes                |
| `git push`                    | Push commits to remote repo   |
| `git pull`                    | Pull updates from remote repo |

---

## 14. Summary

In this demo, we:

* Configured Git (name and email).
* Created a **local repository** in JupyterLab.
* Generated and added **SSH keys** to GitHub.
* Created a **remote repository** on GitHub.
* Linked local and remote repositories.
* Committed and pushed a notebook file.
* Made additional changes and re-pushed them.
