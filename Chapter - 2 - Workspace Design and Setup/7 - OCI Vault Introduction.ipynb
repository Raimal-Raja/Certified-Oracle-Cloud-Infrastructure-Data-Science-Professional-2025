🏛️ What is OCI Vault?

OCI Vault is a managed Oracle service for centrally managing encryption keys and credentials.

It ensures credentials are not stored locally or in code, avoiding common security leaks.

Vault integrates with:

OCI SDK

Command Line Interface (CLI)

API clients

ADS SDK for Data Science workflows

📌 Use Case: Store database credentials, API keys, and service passwords in Vault instead of notebooks or scripts.

🗄️ Key Concepts
1. Vault Types

Virtual Private Vault:

Dedicated, isolated partition in a hardware security module (HSM).

Stores up to 1,000 key versions.

Supports backup to OCI Object Storage (disaster recovery & cross-region replication).

Shared Vault:

Shared HSM partition with other Oracle customers.

Lower cost, but fewer features (e.g., no backup).

2. Keys

Keys are logical entities representing cryptographic material.

Encryption Algorithms Supported: AES, RSA, ECDSA.

Types of Keys:

Master Encryption Keys: Created/imported by you, used to generate data encryption keys.

Data Encryption Keys: Dynamically generated, encrypted with master keys (envelope encryption).

Wrapping Keys: Used to wrap other keys securely.

📌 Key Rotation:

Each key has versions.

Rotating a key generates a new version.

Old versions can decrypt, but not encrypt, reducing risks.

3. Secrets

Secrets = credentials (passwords, tokens, API keys, etc.).

Stored securely in Vault instead of code/config files.

Advantages:

Stronger security than storing in notebooks.

Code becomes more robust (only update Vault, not scripts).

Supports automatic rotation (new versions without breaking code).

📌 Access Control:

Vault uses OCID (Oracle Cloud Identifier) for secrets.

Access controlled via OCI IAM permissions.

Easier to control secret access than notebook access.

⚙️ Workflow for Using OCI Vault

Create a Vault (Shared or Virtual Private).

Generate or import keys (AES, RSA, ECDSA).

Rotate keys periodically to reduce risk.

Create secrets (passwords, tokens, credentials).

Use OCID in code to fetch secrets securely via SDK/CLI.

Rotate secrets when credentials change – Vault automatically handles versioning.

📝 Wrap-Up

OCI Vault securely stores keys and secrets, preventing credentials from leaking through code.

Vault supports:

AES, RSA, and ECDSA encryption.

Two vault types: shared and virtual private.

Envelope encryption with master and data keys.

Secrets are versioned and can be rotated to maintain compliance and security.

Best Practice: Always store credentials in Vault, never in code.
