## Accelerated Data Science (ADS)
## Software Development kit (SDK)

ADS SDK covers the end-to-end life cycle of machine learning models from data acquistion to model evalutions.


### Local Environment Installation
python3 -m pip install oracle-ads

### ADS SDK: Features
connect to data sources, data visualization, feature engineering, model training, model evalutions, model interpretation, model deployement, 

### Connect to data sources
local storage, object storage, oracle database, third-party cloud providers, mongo database and NoSQL database instance, OCI  big Data service, http(s).

### Data Visualization
is part of EDA, helps you understand data sets, Features types classes, provides standard visualizations and allow for customizations.

Provides basic information about a data set:
summary statistic, summary visualization of each feature, correction heat maps.


### Feature Engineering
Transforms existing features into new ones to improve ML models quality

Has built-in functionality and tools to;
turn  a data set into an ADS dataset object, provide recommended transformations to a data set, support categorical encoding, null values and imputation


### Model Training
Model Training using oracle labs'autoML can; automate training multiple algorithms, optimize model hyperparameters, compare performance

ADStuner performs hyperparameter tuning.

after models training, ADS can:
package all necessary files to recreate model (model artificat)

Save the models artifact in the model catalog

### Model Evalution
Allow for comparision between your  trained  ML models with standard metrics,

are offered through ADS as a collection of: tools, metrics, charts

are available through an ADS evalution class that supports, binary classification, multinomail classification, regression

### Model Interpretation and Explainability
is the process that explain/interprets ML models

provides explanation through the ADS module that are: interpretable, model-agnostic what-if, local (explain why the ML made a specific predication)

Globle (explain the general behavior of an ML model)

### Model Deployement
Module provided through ADS; ads.model.framework

uses ADS to deploy common models such as:
oracle  labs autoML, PyTorch, scikit-learn, tensor flow

integrate with OCI logging service, which:
stores access and predications logs, logs uses APIs provided by ADS to assist interactions
