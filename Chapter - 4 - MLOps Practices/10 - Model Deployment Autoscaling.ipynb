## 🎯 Lesson Objective
In this lesson, we explore **autoscaling** in Oracle Cloud Infrastructure (OCI) Data Science.  
Autoscaling helps achieve **cost-efficiency**, **high availability**, and **performance optimization** in model deployments by automatically adjusting compute resources based on demand.

---

## 🧠 Introduction
When setting up a model deployment, choosing the **right compute shape** and **number of instances** is critical. Predicting the ideal scale can be difficult due to changing workloads and external variables.

Autoscaling provides a **dynamic solution** — it automatically scales up or down the number of instances according to real-time demand. This ensures:
- Cost-effective resource utilization  
- Continuous availability  
- Optimal performance for unpredictable workloads

---

## ⚡ What Is Autoscaling?

Autoscaling allows you to define a **range of instances** for a model deployment.  
When demand increases or decreases, the system automatically adjusts the number of active instances within this defined range.

You can also apply autoscaling to **load balancers** by specifying a **bandwidth range** (minimum and maximum). The load balancer automatically adjusts its bandwidth to match incoming traffic, ensuring:
- Enhanced performance  
- Cost control  
- Seamless scaling under fluctuating workloads

## 📊 Metric-Based Autoscaling

**Metric-based autoscaling** is the primary method used in OCI Data Science.  
It triggers scaling actions when a specific **performance metric** (e.g., CPU or memory utilization) meets or exceeds a defined threshold.

### 🔍 How It Works:
1. Performance metrics (like CPU utilization) are continuously **collected by OCI Monitoring Service**.  
2. These metrics are **aggregated and averaged** across all instances in the deployment.  
3. When consecutive metric values meet the scaling threshold, an **autoscaling event** is triggered.  
4. A **cooldown period** follows each event to let the system stabilize before another scaling adjustment.  

Metrics continue to be evaluated even during the cooldown period, ensuring readiness for the next scaling action.

---

## ⚙️ Predefined and Custom Metric Types

OCI offers two types of metric-based scaling policies:

| Type | Description |
|------|--------------|
| **Predefined Type** | Uses built-in metrics like **CPU utilization** or **Memory utilization**. Users set thresholds for scale-up and scale-down events. |
| **Custom Metric Type (NQL Expression)** | Users define custom scaling conditions using **NQL (Monitoring Query Language)**. Any model deployment metric can be used. |

Custom metric expressions allow for:
- Multiple metric combinations (using AND/OR)
- Custom aggregation functions
- Adjustable evaluation windows  
This flexibility enables **fine-grained control** over autoscaling behavior.

## 🧰 Configuring Autoscaling for Model Deployment

You can create a model deployment with **compute and load balancer autoscaling** pre-configured using:
- **OCI Console**
- **OCI CLI**
- **OCI Data Science API**

### Configuration Steps:
1. Define autoscaling **instance range** (min and max).  
2. Choose **metric type** (CPU, memory, or custom NQL).  
3. Set **scaling thresholds** for triggering events.  
4. Configure **cooldown period** to avoid rapid scaling.  
5. Add required **IAM policy** for autoscaling permissions.

---

## 🧩 IAM Policy Requirement

For autoscaling to function correctly, the following policy must be added to your tenancy:

```

Allow service autoscaling to read metrics in tenancy
where target.metrics.namespace='oci_datascience_modeldeploy'

```

This enables the autoscaling service to read and evaluate model deployment metrics.

---

## 🧠 Understanding Metrics in Model Deployments

Metrics are automatically available for all model deployments in the **OCI Data Science Model Deploy namespace** — no additional monitoring setup is needed.

### Common Metric Dimensions:
- `resourceId`
- `statusCode`
- `statusFamily`
- `instanceId`
- `result`
- `networkType`

These metrics can be used to define custom scaling logic.

---

## 🔧 Modifying Autoscaling Configurations

- **Active Deployments:**  
  Only **autoscaling policies** can be modified (not other infrastructure settings).  
  You can still update metadata fields such as:
  - Display name
  - Description
  - Tags

- **Inactive Deployments:**  
  You can modify **all settings simultaneously**, including autoscaling and infrastructure configurations.

## 🧮 Example: Autoscaling Configuration

| Parameter | Description |
|------------|--------------|
| **Metric** | CPU Utilization |
| **Scale-Up Threshold** | > 80% for 5 minutes |
| **Scale-Down Threshold** | < 30% for 10 minutes |
| **Instance Range** | 1 – 5 |
| **Cooldown Period** | 10 minutes |

This setup allows the deployment to automatically add more compute instances when CPU load exceeds 80% and scale back when it drops below 30%.
