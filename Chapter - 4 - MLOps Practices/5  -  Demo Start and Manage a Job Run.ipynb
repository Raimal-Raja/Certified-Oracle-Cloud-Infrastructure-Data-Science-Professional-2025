# 🚀 Starting and Managing Job Runs in Oracle Cloud Infrastructure (OCI) Data Science

## 👋 Introduction
Hello! My name is **Lyudmil Pelov**, and I’m a **Product Manager** in the **Oracle Data Science and AI Services** team.  
In this demo, we’ll learn how to **start and manage job runs** within the **OCI Data Science** environment.

Now that we have our job created, we can **execute one or multiple job runs simultaneously**, allowing for flexible experimentation and testing.

---

## ▶️ Step 1: Starting a Job Run
To start a new job run:

1. Navigate to the **Jobs** section in your OCI Data Science project.
2. Click on the **Start Job Run** button.
3. The **Job Runs Table** displays previous runs, including:
   - Job **status** (e.g., Succeeded, Failed, In Progress)
   - **Lifecycle details**
   - **User** who initiated the run
   - **Timestamp** of when the job was accepted

This overview helps you monitor the status and history of all your job executions.

---

## ⚙️ Step 2: Configuring a New Job Run
When you click **Start Job Run**, a configuration window appears.  
Here, you can modify several key settings before launching your job:

### 🪵 Logging Configuration
- Adjust logging from the default settings established during job creation.  
- Options include:
  - **Disable logging**
  - **Select a new log group**
  - **Choose an existing log** instead of using automatic log creation

### 🌱 Environment Variables
- You can **override** existing variables or **set new ones**.  
- For example:
```

name = Alex

```

### 💬 Command-Line Arguments
- Modify or add arguments to customize execution.  
- Example:
```

-g "Hey"

```

### ⏱️ Maximum Run Time
- Specify how long the job can run before being automatically canceled.
- Example: `100 minutes`

Once configured, click **Start** to launch the job run.

---

## 🔄 Step 3: Job Run Lifecycle
After initiating the job, it transitions through several lifecycle stages:

1. **Accepted** – The job run request is received.  
2. **Provisioning** – OCI begins allocating the necessary infrastructure.  
3. **Running** – The uploaded **artifact (Python code)** is executed.  
4. **Succeeded / Failed** – The job completes, either successfully or with errors.

> 💡 OCI ensures that **you’re billed only for the duration of the job execution**, not while the infrastructure is idle.

You can monitor the progress in real time from the **Job Runs Table**.

---

## 🧠 Step 4: Running Multiple Jobs in Parallel
You do **not** need to wait for one job to complete before starting another.  
For example:
- If you’re training multiple machine learning models with different parameters, you can start **several runs simultaneously**.
- Each run can have **unique environment variables or command-line arguments**.

This feature makes it easy to **experiment and compare** results in parallel.

---

## 🧩 Step 5: Handling Job Failures
If a job run **fails**:
1. Click on the **failed job run**.
2. Open the **log file** to review the detailed error message.
3. Identify the issue (e.g., incorrect variable, missing dependency) and fix it.
4. Restart the job once corrected.

> 🔍 Logs are the best resource for diagnosing failures and improving reliability.

---

## ✏️ Step 6: Editing Job Details
After a job run is completed:
- You can **edit job details** such as:
- The **job name**
- **Tags** for organization
- **Compute shape** or **storage size**

Click **Edit**, make the desired changes, and then **Save** to update your job.

---

## 🧬 Step 7: Cloning a Job Run
OCI provides a powerful **Clone Job Run** feature available via the web interface.

When you clone:
- All previously defined **environment variables** and **command-line arguments** are automatically copied.
- You can modify only the parameters you need to change (for example, updating one variable).

### To clone a job:
1. Click on **Clone** next to an existing job run.
2. Optionally update:
 - Job **name** (e.g., append a timestamp)
 - Any **specific variables or arguments**
3. Click **Clone Job Run** to start a duplicate run with the new configuration.

---

## 🗑️ Step 8: Deleting a Job
To delete a job:

1. Select the job from the list.
2. Click **Delete**.
3. Confirm your action in the pop-up message.

The job and its associated metadata will be permanently removed from the OCI console.

---

## 🛑 Step 9: Canceling a Job Run
You can **cancel a job** while it is running or even during the provisioning stage.

### To cancel:
1. Start a new job run (for example, by cloning an existing successful job).  
2. Once the job enters the **preparing** or **running** state, the **Cancel** button becomes active.  
3. Click **Cancel**, then confirm in the dialog box.

> ✅ Once canceled, **infrastructure is immediately released** and **metering stops**, ensuring you don’t pay for unused runtime.
