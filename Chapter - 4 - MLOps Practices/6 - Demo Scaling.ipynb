# 📘 Demo: Scaling Up and Down Jobs and Notebooks in Oracle Cloud Data Science Service

## 🎯 Objective
In this demo, we will learn how to **scale up and down jobs and notebooks** in **Oracle Cloud Data Science Service**. This capability helps optimize performance and costs by adjusting CPU, memory, and storage resources based on workload needs.

---

**Lyudmil Pelov**, Product Manager – Oracle Data Science and AI Services Team

---

## 🧠 Introduction
Sometimes, after running a job, you may notice that **CPU**, **memory**, or **storage** utilization is very high. In such cases, Oracle Cloud allows you to **edit your job configuration** to allocate more resources.

You can increase:
- CPU cores (OCPUs)
- Memory size
- Storage capacity
- Or even switch to **GPU shapes** for heavy model training workloads.

---

## ⚙️ Step 1: Editing a Job Configuration
To edit your existing job:

1. Navigate to your **Job List**.
2. Select the job you wish to modify.
3. Click on the **Edit** button.

Here, you can modify the **compute shape** used for your job.

- You can select from **Fast Launch Shapes** or **Standard Shapes**.
- For example, if your current shape is `VM.Standard2.1` (the smallest shape), you can switch to a shape with **more OCPUs** and **greater memory**.

---

## 🖥️ Step 2: Changing Compute Shape and Storage
When you click **Change Shape**, you have the ability to:
- Choose a shape with higher performance specifications.
- Select a **GPU-enabled shape** (e.g., `VM.GPU3.1`) for model training.
- Adjust the **storage size** if the previous allocation was insufficient.

Once the configuration changes are made:
- Click **Save Changes**.
- Start a new job run to immediately use the new compute settings.

---

## 📊 Step 3: Scaling Up Notebook Sessions
Scaling up is not limited to jobs — you can also **scale your notebook session**.

To view notebook metrics:
1. Click on your **Notebook session**.
2. Monitor **CPU** and **memory utilization**.
3. If utilization is high, scaling up may be necessary.

---

## ⚠️ Step 4: Deactivating the Notebook Before Scaling
Before scaling up a notebook:
- You **must deactivate it first**.

To do this:
1. Click **Deactivate**.
2. Confirm by selecting **Deactivate** again in the confirmation window.

💡 **Note:**  
Deactivating stops the **billing (metering)** for the compute instance but **does not delete** your block storage.  
Your **files and data remain safe** on the attached block storage.

---

## 🚀 Step 5: Reactivating and Scaling Up
Once the notebook is deactivated:
1. Click **Activate** to restart it.
2. During activation, you can:
   - Select a new **compute shape** (e.g., with more OCPUs or memory).
   - Increase **block storage size**.

You can choose from:
- **Intel-based shapes** (high CPU/memory)
- **Flex shapes**
- **GPU shapes**

For instance, you might select an **Intel shape with higher performance** and also increase the **block storage capacity**.

---

## ✅ Step 6: Activating the Scaled Notebook
After selecting your new configuration:
- Click **Activate**.

Your notebook will start with:
- The **new compute instance shape**
- The **new block storage size**

💡 **Important:**  
All your previous files and data on the block storage remain **intact**, allowing you to **resume work seamlessly** without data loss.

---

## 🏁 Conclusion
Scaling up and down your **jobs** and **notebooks** in Oracle Cloud Data Science Service gives you flexibility to:
- Optimize performance for heavy workloads.
- Reduce costs by deactivating unused sessions.
- Adjust compute and storage resources dynamically as your project evolves.
