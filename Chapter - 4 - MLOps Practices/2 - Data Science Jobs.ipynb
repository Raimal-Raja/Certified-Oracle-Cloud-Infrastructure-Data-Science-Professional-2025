## 🧩 Introduction to OCI Data Science Jobs

**Speaker:** *Lyudmil Pelov, Product Manager, Oracle Data Science and AI Services*

The **Jobs Service** is a core component within **Oracle Cloud Infrastructure (OCI) Data Science**, serving as a foundation for **MLOps** on Oracle Cloud.  
It allows users to define and execute **repeatable, automated tasks** on **fully managed infrastructure** that is provisioned only when needed — optimizing **cost and performance**.

---

## ⚙️ Purpose and Benefits of Jobs

### ✅ Key Capabilities:
- Automates **data preparation**, **model training**, **batch inference**, and other tasks.
- Provides **on-demand compute** — infrastructure is created when needed and deprovisioned afterward.
- Enables **custom task execution** using Python, Bash, or complex multi-file projects.

### 💡 Benefits:
- **Fully managed** – no need to operate or maintain external batch systems.
- **Cost-efficient** – pay only for active job duration.
- **OCI integrated** – leverages databases, networking, and security features.
- **Scalable and flexible** – adjust compute shape or block storage per task.
- **Secure** – supports custom VCNs, IAM policies, and Vault secrets.

---

## 🧱 Core Concepts: Jobs and Job Runs

| **Component** | **Description** |
|----------------|-----------------|
| **Job** | A **template** describing a task — defines compute shape, artifact, and parameters. |
| **Job Run** | A **single execution** of a job, often with customized variables or inputs. |

### 🔹 The Job Artifact:
- Contains **instructions** or **scripts** to be executed.
- **Immutable** after upload — cannot be changed.
- Can include **environment variables** or **CLI arguments**.

### 🔹 The Job Run:
- Executes based on job definitions and parameters.
- Automatically **provisions and deprovisions** infrastructure.
- Supports **monitoring**, **logging**, and **custom configurations** per run.

---

## 🧠 The Job Lifecycle

1. **Create a Job**
   - Define job artifact, compute shape, and environment variables.
2. **Run the Job**
   - Start job runs with optional customized parameters.
3. **Monitor and Log**
   - View compute metrics, logs, and runtime status.
4. **Complete or Cancel**
   - Job deprovisions resources and triggers follow-on processes.

### ⏱️ Important Limits:
- Maximum runtime for a job: **30 days**.
- Automatic termination after timeout.

---

## 🧰 Artifacts and Execution Options

Jobs can execute different **artifact types** depending on project complexity:

| **Artifact Type** | **Description** |
|--------------------|-----------------|
| **Python / Bash File** | For simple, single-script jobs. Python comes preinstalled. |
| **ZIP / TAR Archive** | For complex projects with multiple files or dependencies. |
| **Runtime YAML File** | Defines environment variables or dependencies for ZIP/TAR jobs. |

### 🧭 Environment Control:
- Use **Conda environments** for isolated dependency management.
- Set the **`JOB_RUN_ENTRYPOINT`** variable to specify the main execution file.
- Compute shape and storage can be edited between runs for **scalability**.

## 🚀 Batch Inference with Jobs

**Batch inference** allows asynchronous, large-scale model predictions over batches of data.

### 🧩 Types of Batch Processing:

| **Type** | **Description** | **Use Case Example** |
|-----------|-----------------|----------------------|
| **Regular Batch** | Processes a large dataset periodically. | Daily analytics or prediction jobs. |
| **Mini Batch** | Processes smaller sets more frequently. | Fraud detection every few minutes. |
| **Distributed Batch** | Splits large data into chunks for parallel execution. | Large-scale distributed training or inference. |

### 💡 Tip:
Results can be stored in **Object Storage**, **Databases**, or other OCI services.

---

## 🧮 Example: Mini Batch for Fraud Detection

A **banking application** runs a mini batch every few minutes:
- Fetches recent transaction data.
- Runs an ML model to detect fraudulent activity.
- Triggers alerts for suspicious patterns.

This allows **fast and continuous analysis** rather than waiting for full-day data accumulation.

---

## 🧱 Distributed Batch (Parallel Processing)

- Large datasets are **divided into chunks**.
- Multiple **jobs run simultaneously**, improving performance.
- Each sub-job operates **independently**, ideal for “**embarrassingly parallel**” workloads.

---

## 🧩 Scaling Compute and Storage

To reduce runtime:
- **Increase compute shape** (e.g., from small VM to GPU).
- **Expand block storage** for larger datasets.
- Adjust both **within jobs or notebook sessions**.
