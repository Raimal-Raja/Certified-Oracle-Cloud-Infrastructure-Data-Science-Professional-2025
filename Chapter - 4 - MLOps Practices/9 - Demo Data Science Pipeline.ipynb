# 🧠 Data Science Pipeline Demonstration  
**Instructor:** Hemant Gahankari  
**Role:** Senior Principal Training Lead, Oracle University  

---

## 🎯 Lesson Objective
In this demonstration, we will **learn how to create and run a data science pipeline** in Oracle Cloud Infrastructure (OCI) to **automate an end-to-end machine learning workflow**.  
The demonstration walks through creating a pipeline with multiple steps, configuring each step, and monitoring the pipeline execution to deploy the best-performing model.

---

## 🧩 Step 1: Setting up the Environment

The code for the pipeline steps is available in the **Oracle Samples GitHub repository** under  
**`OCI Data Science AI Samples`**.

### 📁 Repository Access
You can download the code by:
1. Clicking the **Code** button on GitHub.
2. Selecting **Download ZIP**.

After downloading and extracting:
- The **master folder** contains a **`pipelines/samples/employee-attrition`** directory.
- Inside this directory are **multiple ZIP files**, each representing a specific step in the pipeline.

---

## 🧱 Step 2: Creating a Data Science Project

To begin, create a **Data Science Project** in OCI.  
In this demo, the instructor already has a project set up.

### 🔧 Navigate to:
`Data Science → Projects → Pipelines → Create Pipeline`

Provide the following:
- **Pipeline Name:** `Employee Attrition Pipeline`  
- **Description:** `Demo pipeline for end-to-end ML workflow`

---

## ⚙️ Step 3: Configuring the Pipeline

Define the **default configuration** for the pipeline:

| Setting | Description |
|----------|--------------|
| **Environment Variable** | `data_location` → points to OCI Object Storage bucket (`pipeline-temp-bucket`) |
| **Compute Shape** | `VM.Standard2.2` |
| **Block Volume Size** | 50 GB |
| **Logging Group** | Pre-created log group (auto-log creation enabled) |

After setting these defaults, the pipeline is ready to define individual **pipeline steps**.

---

## 🔄 Step 4: Adding Pipeline Steps

Click **Add Pipeline Step** and configure each step as shown below.

### 🔹 Step 1 – Data Processing
- **Type:** Build by Script  
- **Step Name:** `step1`  
- **Description:** Data Processing  
- **Depends on:** None  
- **Artifact:** Upload ZIP file (`dataproc.zip`)  
- **Entry Point:** Python file inside the ZIP  
- **Action:** Save  

---

### 🔹 Steps 2a, 2b, and 2c – Model Training (Parallel Execution)
Each step trains a model using a different algorithm.

| Step | Description | Depends On | Artifact | Entry Point |
|------|--------------|-------------|-----------|--------------|
| Step 2a | Train Linear Regression Model | Step 1 | `lr.zip` | `lr.py` |
| Step 2b | Train Random Forest Model | Step 1 | `rf.zip` | `rf.py` |
| Step 2c | Train XGBoost Model | Step 1 | `xgb.zip` | `xgb.py` |

All three steps run **in parallel** after Step 1 completes.

---

### 🔹 Step 3 – Evaluate and Deploy Model
- **Step Name:** `step3`  
- **Description:** Evaluate and Deploy  
- **Depends on:** Steps 2a, 2b, 2c  
- **Artifact:** `evaluate_and_deploy.zip`  
- **Entry Point:** `evaluate_and_deploy.py`  
- **Action:** Save  

---

## 📊 Step 5: Reviewing the Pipeline

After creation, the pipeline visually displays all steps and dependencies:

| Step | Depends On | Type |
|------|-------------|------|
| Step 1 | None | Script |
| Step 2a | Step 1 | Script |
| Step 2b | Step 1 | Script |
| Step 2c | Step 1 | Script |
| Step 3 | Steps 2a, 2b, 2c | Script |

You can now verify:
- **General Information**
- **Infrastructure Configuration**
- **Logging Configuration**
- **Default Environment Variables**

---

## 🚀 Step 6: Running the Pipeline

Click **Start Pipeline Run** to begin execution.

### Run Settings
- **Run Name:** `Run-11` (or any custom name)
- **Override Configuration:** No
- **Environment:** Add `CONDA_ENV_SLUG` variable for Conda environment setup

Repeat environment configuration for **Steps 1, 2a, 2b, 2c, and 3**.

---

## 🧠 Step 7: Monitoring Execution

You can monitor real-time pipeline status:

| Step | State | Description |
|------|--------|-------------|
| Step 1 | Accepted → Running → Succeeded | Data processed |
| Step 2a | Waiting → Accepted → Succeeded | Linear Regression trained |
| Step 2b | Waiting → In Progress → Succeeded | Random Forest trained |
| Step 2c | Waiting → In Progress → Succeeded | XGBoost trained |
| Step 3 | Waiting → Accepted → Succeeded | Evaluated & Deployed model |

All steps run successfully, as indicated by their final **Succeeded** state.

---

## 📈 Step 8: Reviewing Logs and Model Results

The **AUC scores** for the models (from logs):

| Model | AUC Score |
|--------|------------|
| Linear Regression | 0.85 |
| Random Forest | 0.81 |
| XGBoost | 0.837 |

✅ **Best Model:** Linear Regression (AUC = 0.85)  
This model was **automatically deployed** after evaluation.

---

Navigate to:
`Data Science → Projects → Models`

You can see:
- The best-performing **model selected**
- The corresponding **model deployment**
