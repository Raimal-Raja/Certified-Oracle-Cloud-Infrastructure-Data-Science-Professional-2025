## 🎯 Objective
In this module, we will learn about the **monitoring and logging capabilities** available for jobs in **Oracle Cloud Data Science Service**.  
Monitoring and logging represent the **final stage** in the job lifecycle — just before the infrastructure is de-provisioned — and are essential for **tracking performance**, **analyzing metrics**, and **debugging job runs**.

---

## 🧠 Introduction
Monitoring and logging give you deep insights into your job’s **performance**, **resource usage**, and **execution details**.  
They also generate **records for later reference**, helping you improve and optimize future job runs.

- **Monitoring** includes metrics and alarms.  
- **Logging** captures job outputs for debugging and auditing.  

Together, they form a complete observability solution for Oracle Data Science jobs.

---

## 📊 Step 1: Understanding Monitoring
Monitoring in Oracle Cloud consists of **metrics** and **alarms**, which allow you to:
- Check the **health**, **capacity**, and **performance** of cloud resources.
- Use data to determine **when to scale resources**, **troubleshoot issues**, or **analyze system behavior**.

### 🔹 Metrics
Oracle services like **Data Science**, **Compute**, and **Databases** continuously emit **real-time metrics** to the Oracle Monitoring service.

Common metrics for job runs include:
- **CPU Utilization (%)**
- **GPU Utilization (%)**
- **Memory Usage**
- **Network Traffic (Bytes In/Out)**
- **Disk Utilization**

When these metrics reach certain thresholds, you may need to **scale up** compute shapes or **increase storage** to handle the workload effectively.

---

## ⏰ Step 2: Alarms and Notifications
### 🔸 Alarms
An **alarm** is a rule that triggers a **notification** when a metric crosses a specified threshold.  
For example:
> An alarm can be configured to trigger a notification when CPU utilization exceeds **80%**.

When the threshold is breached:
1. The alarm triggers an **event**.
2. A **notification** is sent through subscribed channels like:
   - **Email**
   - **SMS**
   - **Slack**
   - **Other notification topics**

### 🔸 Integration
Metrics, alarms, and notifications work together to provide a **proactive monitoring system** that helps you:
- Detect performance issues early.
- Take automated actions (e.g., scaling or restarting).
- Stay informed about system health in real-time.

## 🧾 Step 3: Job Logging Options
With Oracle Data Science jobs, you can use **service logs** or **custom logs** to capture job activity.

### 🟩 Service Logs
- These are **standard logs** automatically emitted by the job run to the **OCI Logging Service**.
- Both **standard output (stdout)** and **standard error (stderr)** from your job artifact are captured.
- The **resource principal** running the job must have appropriate permissions to write to the logs.

### 🟨 Custom Logs
- Allow you to **define your own logging context** and **choose where to store log data**.
- Useful for **application-specific debugging** or **custom monitoring setups**.

### 🛠️ Enabling Logging
Logging is **optional but highly recommended** for:
- Debugging failed jobs.
- Monitoring real-time job progress.

You can:
- Provide your **own log group** and **log**.
- Enable **automatic log creation** by the Data Science service at the start of each job run.

When enabled, each job run can:
- Use its **own log** or share logs with other runs.
- Automatically store outputs within your specified log group.

💡 **Note:** Logs are **not deleted** when the job or job runs are removed, ensuring you retain full historical visibility.

---

## 🪶 Step 4: Event Service Integration
The **Event Service** in Oracle Cloud allows you to respond to **resource changes** such as job or job run events.

### 📩 Events
An **event** is a structured message that signals a **change** in a resource’s state — typically a CRUD (Create, Read, Update, Delete) operation.

### ⚙️ Rules and Actions
- Users create **rules** to specify which events to monitor.  
- Each rule can trigger one or multiple **actions**.

Possible actions include:
- **Triggering notifications**
- **Invoking Oracle Functions**
- **Streaming event data**

The Event Service ensures **at least one guaranteed delivery** of every triggered action.

---

## 🔄 Step 5: Example Workflow
Here’s how these services interact during a job lifecycle:

1. **Job starts →** Monitoring begins collecting metrics (CPU, memory, etc.).
2. **Metrics exceed threshold →** Alarm triggers.
3. **Alarm triggers →** Notification is sent via email/SMS.
4. **Job logs →** Captured in OCI Logging Service for review.
5. **Job updates →** Event Service triggers an automated function or message.
6. **Job finishes →** Logs and metrics remain for analysis and auditing.
