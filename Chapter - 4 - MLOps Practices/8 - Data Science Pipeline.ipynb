## 🧠 Introduction
In the earlier lesson on **Data Science Jobs**, we learned how to implement custom tasks using job artifacts.  
Now, with **Pipelines**, we can design and execute **end-to-end workflows** that consist of multiple tasks connected together.

A **Data Science Pipeline** is a new feature of the **Oracle Cloud Infrastructure (OCI) Data Science Service**.  
It allows data scientists to automate entire workflows — from data processing to model deployment — in a controlled and repeatable way.

---

## 🔹 Step 1: Understanding the Data Science Pipeline

A **pipeline** consists of one or more **steps**.  
Each step performs a discrete operation such as:

- Data processing  
- Model training  
- Model evaluation  
- Model deployment  

Pipelines are flexible — you can mix **different environments** and **programming languages** within a single pipeline.  

For example:
> You can use **Python** for data preprocessing and **Java** for model training — both within the same pipeline.

---

## ⚙️ Step 2: Workflow and Dependencies

Each **pipeline step** can depend on other steps to form a logical workflow.

- Steps can execute **sequentially** (one after another) or **in parallel**.
- By default, all steps attempt to execute in parallel unless dependencies are explicitly defined.

### 🧩 Example Workflow:
| Step | Task | Description |
|------|------|--------------|
| Step 1 | Data Processing | Cleans and prepares data |
| Step 2a–2c | Model Training | Trains multiple models (e.g., Linear Regression, Random Forest, XGBoost) |
| Step 3 | Model Evaluation | Compares model performance |
| Step 4 | Model Deployment | Deploys the best-performing model |

Defining dependencies ensures that:
- Data processing occurs before model training.  
- Model evaluation occurs after training.  
- Deployment occurs only after successful evaluation.

## 🧰 Step 3: Pipeline Configuration

A pipeline is a **resource** that defines how the workflow executes.  
To configure it, you need to define:

1. **Compute Shape** – Determines the hardware resources (OCPUs, GPUs, and memory).  
2. **Block Volume** – Storage capacity for intermediate files and datasets.  
3. **Environment Variables** – Used to pass data such as file locations or runtime parameters.  
4. **Logging Settings** – To capture job outputs and debug information.  
5. **Maximum Runtime** – Defines time limits for pipeline execution.

### 💡 Default Settings
- Pipelines have **default configurations** that apply to all steps.  
- However, each **pipeline step** can override these defaults with its own configuration.

---

## 🧩 Step 4: Pipeline Steps in Detail

A **pipeline step** represents a single executable task.  
Each step has its **own configuration**, which can override the pipeline’s default settings.

Pipeline steps can be implemented in two ways:

### 🟢 Option 1: Script-Based Step
- Implemented using a **Python**, **Bash**, or **Java** script.  
- The step artifact can be:
  - A single script file, or  
  - A compressed ZIP file containing multiple files.

### 🟣 Option 2: Job-Based Step
- Uses a **pre-created Oracle Data Science job** identified by its **OCID (Oracle Cloud Identifier)**.  
- Enables reuse of existing jobs as part of the pipeline.

### 🔗 Resource Access
Pipeline steps can access:
- **OCI Resources** such as Object Storage, Model Catalog, and Databases (with correct IAM policies).  
- **External resources** through custom **VCN (Virtual Cloud Network)** configurations.

---

## 🔄 Step 5: Pipeline Lifecycle

Every pipeline follows a defined lifecycle:

| Stage | Description |
|--------|-------------|
| **Creating** | The pipeline is being initialized. |
| **Active** | The pipeline is ready for execution. |
| **Pipeline Run** | Represents an individual execution instance of the pipeline. Multiple runs can exist. |
| **Deleted** | Once the pipeline is no longer required, it can be removed permanently. |

Each **pipeline run** executes according to its configuration, logging outputs, and generating metrics for analysis.

## 🧪 Step 6: Demonstration Scenario

Let’s review the scenario used to record the demonstration video.

### 📍 Step 1: Data Processing
- Reads input data from **OCI Object Storage**.  
- Performs operations like:
  - Dropping unnecessary columns  
  - Applying **label encoding**  
  - **Scaling** data  
  - Splitting data into **train** and **test** sets  
- Saves the processed data back to **Object Storage**.  
- Sets **pipeline variables** to communicate data locations to the next steps.

### ⚙️ Step 2: Model Training
- Trains **three models** using different algorithms:
  - Linear Regression  
  - Random Forest  
  - XGBoost  
- Stores all trained models in the **OCI Model Catalog**.

### 🧠 Step 3: Model Evaluation and Deployment
- Fetches all models from the Model Catalog.  
- Evaluates their performance metrics.  
- Selects the **best-performing model**.  
- Deploys the chosen model to production.
