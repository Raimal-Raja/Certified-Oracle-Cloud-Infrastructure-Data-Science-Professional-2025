## ⚙️ What is MLOps?

**MLOps** stands for **Machine Learning Operations**.  
It is the application of **DevOps practices** to **machine learning workflows** to improve:
- **Efficiency**,  
- **Quality**, and  
- **Consistency** of machine learning solutions.  

MLOps focuses on the **standardization**, **streamlining**, and **automation** of the **machine learning lifecycle** — from data collection and model training to deployment and monitoring.

In MLOps, machine learning assets such as datasets, models, and pipelines are treated like any other **software asset** within a **Continuous Integration and Continuous Delivery (CI/CD)** environment.

---

## 🧩 Core Principles of MLOps

MLOps is built upon several **key principles** adapted from DevOps:

1. **Standardization of ML Workflows**  
   Ensures consistent and repeatable training and deployment practices.

2. **Automation of ML Pipelines**  
   Reduces manual intervention in tasks like data preprocessing, model training, and testing.

3. **Continuous Integration (CI)**  
   Automates the integration and validation of new data and models.

4. **Continuous Deployment (CD)**  
   Enables automatic release of validated models into production.

5. **Continuous Training (CT)**  
   Unique to MLOps — ensures that models are **retrained on new data** to maintain accuracy and relevance.

---

## 🔄 Relationship Between DevOps and MLOps

MLOps borrows heavily from **DevOps**, but extends it with **data-centric processes**.

| DevOps Focus | MLOps Focus |
|---------------|--------------|
| Software code versioning | Data and model versioning |
| Continuous Integration / Deployment | Continuous Integration / Deployment / Training |
| Static application updates | Dynamic model retraining |
| Monitoring software health | Monitoring model performance and data drift |

While **DevOps** automates software builds, testing, and deployments,  
**MLOps** additionally automates **model retraining and evaluation** based on **new data inputs**.

---

## 📈 The Need for Continuous Training

One key difference between DevOps and MLOps is **data dynamics**.

- **Software** tends to be static once deployed.  
- **Machine learning models**, however, depend on **ever-changing data**.  

When the data distribution changes (a phenomenon known as **data drift**), model performance may degrade.  
To maintain accuracy, models must be **continuously retrained** and **validated**.

This is why **Continuous Training (CT)** is essential in MLOps:
- Retrains models as new data arrives.  
- Validates model performance before redeployment.  
- Ensures high-quality predictions in real-world environments.

## 🏗️ MLOps Lifecycle Stages

The **MLOps workflow** evolves as an organization matures in automation.  
There are **three key stages** of MLOps development:

---

### 1. Manual ML Process (Experimental Stage)

- Early, **experimental phase** using tools like **Jupyter Notebooks**.  
- Data preparation, model training, and validation are performed **manually**.  
- Serves as the foundation for automation.

**Example tools:**  
Jupyter Notebook, ADS SDK, Pandas, Scikit-learn.

---

### 2. Automated ML Pipeline

- The **training and validation processes** become **automated**.  
- New incoming data triggers **automatic retraining** and **evaluation**.  
- Ensures models are **optimized continuously** with minimal manual effort.

**Key benefit:** Reduces training time and improves model freshness.

---

### 3. Full CI/CD Pipeline Automation (True MLOps)

- The final stage of automation maturity.  
- **Data ingestion**, **model training**, **testing**, and **deployment** are fully automated.  
- Models are continuously integrated into production environments through triggers such as:
  - New data availability  
  - Monitoring events  
  - Scheduled intervals  

**Continuous Monitoring** ensures that:
- Production data and model performance are constantly tracked.  
- Business metrics are aligned with model performance.  
- Automatic retraining is triggered if model accuracy degrades.

## 🧠 MLOps Architecture in Oracle Cloud

The following outlines the **end-to-end MLOps architecture** in **OCI Data Science**:

1. **Data Ingestion**  
   - Data from business systems or scientists is collected.  
   - Input is sent to **Jupyter Notebooks** for model development.

2. **Model Development**  
   - Data scientists use OCI Notebooks to train and evaluate models.  
   - Models and notebooks are version-controlled.

3. **CI/CD Pipeline with OCI DevOps**  
   - New data or updated notebooks trigger the **DevOps build pipeline**.  
   - The build pipeline executes training jobs and stores model outputs in the **OCI Model Catalog**.

4. **Model Deployment Pipeline**  
   - The **deployment pipeline** is triggered automatically after the build pipeline completes.  
   - The trained model is deployed to a **testing endpoint**.

5. **Model Validation**  
   - The deployed model is validated internally.  
   - Once approved, it is promoted to **production deployment**.

6. **End-User Application**  
   - The final model is integrated into applications for **inference** and **decision-making**.  
   - Continuous feedback loops ensure ongoing improvement.
