# 🧠 Creating and Managing Jobs in Oracle Cloud Infrastructure (OCI) Data Science

## 🎯 Objective
In this demo, we will learn how to **create and manage jobs** in Oracle Cloud Infrastructure (OCI) Data Science. You will see how to configure, upload, and monitor a job artifact to run Python scripts or projects efficiently.

---

## 🧩 Step 1: Accessing the Data Science Service
1. **Log in** to your Oracle Cloud tenancy.  
2. From the **hamburger menu** (☰) on the top left, navigate to:  
   **Analytics and AI → Machine Learning → Data Science.**  
3. If you don’t have a project, create one by clicking **“Create Project.”**  
   - Provide a **Project Name** and **Description**.  
   - Click **Create** to finalize.  

> 🧠 In this demo, we’ll use an existing project to create our job.

---

## ⚙️ Step 2: Creating a Job
1. Inside your project, you’ll see several resource options:
   - **Notebook Sessions**
   - **Jobs**
   - **Model Catalog**
   - **Model Deployments**
2. Select **Jobs** → click on **Create Job.**
3. In the **Create Job** screen:
   - Optionally, change the **Compartment**.
   - Provide a **Job Name** and **Description**.

---

## 📦 Step 3: Uploading the Job Artifact
1. Click **Select a File** to upload your **Python file or ZIP artifact** created earlier.  
2. Select the file from your computer and click **Open**.  
3. OCI console allows uploading files **up to 100 MB** via UI.  
   - For larger files, use the **OCI SDK** for uploading artifacts.

---

## 🧮 Step 4: Configuring Environment Variables and Arguments
1. Set **Environment Variables** expected by your code.  
   - Example:  
     - Variable: `name`  
     - Value: `John`
2. Add **Command-Line Arguments** expected by your script.  
   - Example:  
     - Argument: `-g`  
     - Value: `Hey`
3. Set the **Maximum Runtime (minutes)** — e.g., `100`.  
   This defines how long a job can run before automatic cancellation.

---

## 🖥️ Step 5: Selecting Compute Shape
OCI supports two types of compute shapes:
- **Fast Launch Shapes:**  
  - Pre-warmed instances for quicker startup.  
  - Ideal for small to medium workloads.
- **Custom Configuration:**  
  - Choose **GPU** or **Intel** shapes.  
  - Use for specialized or high-performance tasks.

> 💡 Fast Launch shapes are limited to pre-selected types but can be expanded based on demand.

---

## 🧾 Step 6: Enabling Logging (Optional but Recommended)
1. Enable logging to automatically send logs to **OCI Logging Service**.  
2. Click **Select**, choose a **Log Group**, and enable one of the following:
   - **Automatic log creation** for each job run.
   - **Use existing log** previously created.
3. Click **Select** to confirm your logging setup.

---

## 💾 Step 7: Configuring Storage
- This determines the **block storage volume** attached to your job.  
- Ensure that the **storage size** matches your data and processing needs.

---

## 🌐 Step 8: Network Configuration
You can choose between:
1. **Default Networking Resources** — automatically managed by OCI for standard access.
2. **Custom Network Configuration** — manually select:
   - Virtual Cloud Network (VCN)
   - Subnet  
> 🧠 In this demo, we will use **default networking**, as no special configuration is required.

---

## 🚀 Step 9: Finalizing Job Creation
1. Review all configurations.  
2. Click **Create** to generate your job.  
3. Once created, review details such as:
   - **General Information**
   - **Job Artifact**
   - **Logging and Infrastructure Configuration**

> 📋 Note: The job is only defined at this stage — it’s **not running yet**.  
You’ll need to **execute a Job Run** to start it.

---

## 🧱 Step 10: Reviewing and Managing the Job
- Check **Command-Line Arguments**, **Environment Variables**, and **Runtime Settings**.
- You can:
  - **Edit** job details (name, compute shape, storage, etc.).
  - **Download Artifact** → click *Download Artifact*, choose save location, and click **Save**.
  - **Move Resource** to another compartment.
  - **Add Tags** for organization.
  - **Delete Job** if no longer needed.
