## 🧩 Introduction

In this demo, we’ll learn how to **create a job artifact** that can be used within OCI Data Science Jobs.

A **job artifact** is the file or collection of files that define what your job will execute.  
It can be:

- A single **Python file** containing your code.  
- A **shell script** file.  
- A compressed **ZIP file** containing a full Python project.

There are **no special requirements** for how to write your code — standard Python or Bash scripts are sufficient.

---

## 🧱 Step 1: Starting a Simple Job Artifact

We’ll begin by creating a minimal Python file with just two print statements to signal when the job starts and finishes.

```python
print("Job is starting...")
print("Job is finished.")
````

This is our **base version**, which we’ll extend in the next steps to include timestamps, environment variables, and command-line arguments.

---

## 🕒 Step 2: Adding a Timestamp

We modify the script to print the **current timestamp** when the job begins.

```python
from datetime import datetime

print(f"Job started at: {datetime.now()}")
print("Job is finished.")
```

This allows us to track when each job run begins.

---

## 🌍 Step 3: Reading Environment Variables

Jobs can pass **environment variables**, such as configuration settings or parameters.

Here’s how to read them using Python’s built-in `os` module:

```python
import os
from datetime import datetime

name = os.getenv("NAME", "Unknown User")

print(f"Job started at: {datetime.now()}")
print(f"Hello, {name}!")
print("Job is finished.")
```

If the environment variable `NAME` is provided when the job runs, it will print a personalized greeting.

---

## 💬 Step 4: Reading Command-Line Arguments

We can also pass parameters using **command-line arguments**.
This is done using Python’s `argparse` module.

```python
import argparse
import os
from datetime import datetime

parser = argparse.ArgumentParser()
parser.add_argument("-g", "--greeting", help="Custom greeting message", default="Hello")
args = parser.parse_args()

name = os.getenv("NAME", "Unknown User")

print(f"Job started at: {datetime.now()}")
print(f"{args.greeting}, {name}!")
print("Job is finished.")
```

This allows dynamic input during job execution.
For example:

```bash
python job_artifact.py -g "Welcome"
```

---

## 🧠 Step 5: Integrating with OCI SDK

Now let’s enhance the script to use the **OCI Python SDK**, which lets the job interact with Oracle Cloud resources.

The **Resource Principal** is OCI’s authentication mechanism that allows secure access between services — without needing to store credentials manually.

Below is an extended version that includes **OCI SDK initialization** and **Resource Principal setup**.

```python
import oci
import argparse
import os
from datetime import datetime

def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("-g", "--greeting", help="Custom greeting message", default="Hello")
    args = parser.parse_args()

    name = os.getenv("NAME", "Unknown User")

    print(f"Job started at: {datetime.now()}")
    print(f"{args.greeting}, {name}!")

    # Detect if running locally or as a Job in OCI
    try:
        signer = oci.auth.signers.get_resource_principals_signer()
        print("Running inside OCI Data Science environment using Resource Principal authentication.")
    except Exception:
        print("Running locally without Resource Principal authentication.")
        signer = None

    # Example OCI SDK usage (list compartments)
    if signer:
        identity_client = oci.identity.IdentityClient(config={}, signer=signer)
        user = identity_client.get_user(signer.tenancy_id).data
        print(f"Authenticated OCI user: {user.name}")

    print("Job is finished.")

if __name__ == "__main__":
    main()
```

---

## 🧪 Step 6: Testing Locally

You can run the code locally to ensure everything works before uploading it as a job artifact.

Example commands:

```bash
export NAME="Alice"
python job_artifact.py -g "Hi"
```

Expected output:

```
Job started at: 2025-10-05 14:25:12
Hi, Alice!
Running locally without Resource Principal authentication.
Job is finished.
```

---

## ☁️ Step 7: Using the Artifact in OCI Data Science

Once tested:

1. **Upload** this `.py` file or a `.zip` project as your **job artifact**.
2. Define environment variables and command-line arguments in your **OCI Job settings**.
3. Select your compute shape and click **Run Job**.
4. Monitor logs and outputs directly from the OCI console.

---

## 🧩 Recap

* A **Job Artifact** defines the executable code for your OCI Job.
* It can be **Python, Bash, or ZIP archive** format.
* You can pass parameters via **environment variables** or **CLI arguments**.
* **Resource Principal** provides secure, credential-free OCI access.
* Test your script **locally** before deploying to OCI.

---

## 🧠 Key Takeaway

> OCI Data Science Job Artifacts offer a flexible, language-agnostic way to run repeatable, secure, and scalable workloads — from simple scripts to complete machine learning pipelines — all managed through OCI’s powerful MLOps infrastructure.

---

````
